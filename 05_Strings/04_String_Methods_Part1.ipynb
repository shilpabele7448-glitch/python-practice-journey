## What are String Methods?

String methods are built-in functions that perform different operations on strings such as converting case, removing spaces, checking text, and formatting strings.

## 1. upper()

Converts all characters to uppercase.

```python
text = "python programming"

print(text.upper())
```

## 2. lower()

Converts all characters to lowercase.

```python
text = "PYTHON PROGRAMMING"

print(text.lower())
```

## 3. capitalize()

Capitalizes only the first character.

```python
text = "python programming"

print(text.capitalize())
```

## 4. title()

Converts the first letter of every word to uppercase.

```python
text = "python programming language"

print(text.title())
```

## 5. swapcase()

Converts uppercase letters to lowercase and lowercase letters to uppercase.

```python
text = "Python Programming"

print(text.swapcase())
```

## 6. strip()

Removes spaces from both the beginning and the end of a string.

```python
text = "   Python Programming   "

print(text.strip())
```

## 7. lstrip()

Removes spaces from the left side of the string.

```python
text = "     Python"

print(text.lstrip())
```

## 8. rstrip()

Removes spaces from the right side of the string.

```python
text = "Python     "

print(text.rstrip())
```
