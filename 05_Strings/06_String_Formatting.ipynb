## What is String Formatting?

String formatting allows us to combine text and variables into a single string.

Python supports:

- f-strings (Recommended)
- format() method
- % formatting (Older method)

## 1. f-Strings (Recommended)

f-strings were introduced in Python 3.6.

They are easy to read and are the most commonly used method today.

```python
name = "Shilpa"
age = 22

print(f"My name is {name} and I am {age} years old.")
```

## 2. format() Method

The format() method inserts values into placeholders.

```python
name = "Shilpa"
city = "Pune"

print("My name is {} and I live in {}.".format(name, city))
```

## 3. % Formatting (Older Method)

This is an older style of formatting. It is still found in some legacy Python programs.

```python
name = "Shilpa"
marks = 89

print("Name: %s" % name)
print("Marks: %d" % marks)
```

## Formatting Decimal Numbers

```python
price = 249.9876

print(f"Price: {price:.2f}")
```

## Formatting Multiple Values

```python
name = "Rahul"
course = "Python"
fees = 15000

print(f"Student: {name}")
print(f"Course: {course}")
print(f"Fees: ${fees}")
```

## Formatting with Alignment

```python
print(f"{'Name':<10}{'Marks':>10}")

print(f"{'Amit':<10}{90:>10}")
print(f"{'Neha':<10}{85:>10}")
print(f"{'Riya':<10}{92:>10}")
```

## Formatting Percentages

```python
score = 0.875

print(f"Percentage: {score:.1%}")
```

## Common Mistakes

- Forgetting the `f` before an f-string.
- Using the wrong placeholder in `%` formatting.
- Forgetting curly braces `{}` in f-strings.
