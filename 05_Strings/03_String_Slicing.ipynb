# String Slicing

## Topic

String Slicing

## Description

String slicing allows us to extract a portion (substring) of a string using indexes.

It is one of the most useful features of Python strings.

**Author:** Shilpa

## What is String Slicing?

Slicing means extracting a part of a string.

### Syntax

```python
string[start : stop : step]
```

- **start** → Starting index (included)
- **stop** → Ending index (excluded)
- **step** → Number of positions to move

## Example String

```
P  Y  T  H  O  N
0  1  2  3  4  5
```

````python
text = "PYTHON"

print(text[0:3])
````

````python
text = "PYTHON"

print(text[2:6])
````

````python
text = "PYTHON"

print(text[:4])
````

````python
text = "PYTHON"

print(text[2:])
````

````python
text = "PYTHON"

print(text[:])
````

## Using Step

````python
text = "PYTHON"

print(text[::2])
````

````python
text = "PYTHON"

print(text[1::2])
````

## Reverse a String

````python
text = "PYTHON"

print(text[::-1])
````

````python
name = "Shilpa"

print(name[1:5])
````

````python
city = "Pune"

print(city[-4:-1])
````

````python
sentence = "Python Programming"

print(sentence[7:])
````

## Real-Life Applications

String slicing is useful for:

- Extracting first name
- Extracting last name
- Getting file extensions
- Reading parts of an ID
- Processing text data

## Data Analytics Connection

Slicing helps in:

- Extracting customer IDs
- Cleaning text columns
- Separating date values
- Processing product codes

## Common Mistakes

- Forgetting that the stop index is excluded.
- Using an invalid step value.
- Mixing indexing and slicing.

## Key Points

- Slicing extracts part of a string.
- Start is included.
- Stop is excluded.
- Step controls movement.
- `[::-1]` reverses a string.
