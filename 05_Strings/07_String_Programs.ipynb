## Program 1 - Reverse a String

```python
text = input("Enter a string: ")

print("Reversed String:", text[::-1])
```

## Program 2 - Check Palindrome

```python
text = input("Enter a string: ")

if text == text[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

## Program 3 - Count Vowels

```python
text = input("Enter a string: ")

count = 0

for ch in text.lower():
    if ch in "aeiou":
        count += 1

print("Total Vowels:", count)
```

## Program 4 - Count Consonants

```python
text = input("Enter a string: ")

count = 0

for ch in text.lower():
    if ch.isalpha() and ch not in "aeiou":
        count += 1

print("Total Consonants:", count)
```

## Program 5 - Count Words

```python
sentence = input("Enter a sentence: ")

words = sentence.split()

print("Total Words:", len(words))
```

## Program 6 - Count Characters

```python
text = input("Enter a string: ")

print("Total Characters:", len(text))
```

## Program 7 - Remove Spaces

```python
text = input("Enter a string: ")

print(text.replace(" ", ""))
```

## Program 8 - Convert to Uppercase

```python
text = input("Enter a string: ")

print(text.upper())
```

## Program 9 - Convert to Lowercase

```python
text = input("Enter a string: ")

print(text.lower())
```

## Program 10 - Replace a Word

```python
sentence = input("Enter a sentence: ")

old_word = input("Word to replace: ")
new_word = input("New word: ")

print(sentence.replace(old_word, new_word))
```

## Program 11 - Find Character Frequency

```python
text = input("Enter a string: ")

character = input("Enter a character: ")

print("Frequency:", text.count(character))
```

## Program 12 - Check if String Starts With a Prefix

```python
text = input("Enter a string: ")

prefix = input("Enter prefix: ")

print(text.startswith(prefix))
```

## Program 13 - Check if String Ends With a Suffix

```python
text = input("Enter a string: ")

suffix = input("Enter suffix: ")

print(text.endswith(suffix))
```

## Program 14 - Check if Input Contains Only Digits

```python
text = input("Enter a value: ")

if text.isdigit():
    print("Contains only digits")
else:
    print("Contains other characters")
```

## Program 15 - Count Occurrences of a Word

```python
sentence = input("Enter a sentence: ")

word = input("Enter the word: ")

print("Occurrences:", sentence.lower().count(word.lower()))
```
