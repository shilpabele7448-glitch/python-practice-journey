## 1. Basic for Loop

The simplest way to loop through a tuple is using a `for` loop.

````python
fruits = ("Apple", "Banana", "Mango", "Orange")

for fruit in fruits:
    print(fruit)
````

## How Does the Loop Work?

For every iteration, Python takes one element from the tuple and stores it in the loop variable.

```text
fruits
   ↓
Apple   → fruit
Banana  → fruit
Mango   → fruit
Orange  → fruit

## 2. Loop Using Index

We can use `range()` and `len()` to access tuple elements by their indexes.

````python
numbers = (10, 20, 30, 40, 50)

for i in range(len(numbers)):
    print(numbers[i])
````

## Display Index and Value

Using indexing, we can display both the position and the value.

````python
numbers = (10, 20, 30, 40, 50)

for i in range(len(numbers)):
    print("Index:", i, "Value:", numbers[i])
````

## 3. Using enumerate()

`enumerate()` gives us both:

- Index
- Value

### Syntax

```python
for index, value in enumerate(tuple):
    ...

````python
fruits = ("Apple", "Banana", "Mango", "Orange")

for index, fruit in enumerate(fruits):
    print(index, fruit)
````

## Starting enumerate() from a Different Number

By default, `enumerate()` starts from `0`.

We can change the starting number using `start=`.

````python
fruits = ("Apple", "Banana", "Mango")

for number, fruit in enumerate(fruits, start=1):
    print(number, fruit)
````

## 4. Find Even Numbers

We can combine a loop with an `if` condition.

````python
numbers = (10, 15, 20, 25, 30, 35)

for number in numbers:
    if number % 2 == 0:
        print(number)
````

## Find Odd Numbers

````python
numbers = (10, 15, 20, 25, 30, 35)

for number in numbers:
    if number % 2 != 0:
        print(number)
````

## Find Values Greater Than a Number

````python
numbers = (10, 25, 40, 15, 60, 30)

for number in numbers:
    if number > 30:
        print(number)
````

## 5. Count Positive and Negative Numbers

````python
numbers = (10, -5, 20, -10, 30, -15)

positive = 0
negative = 0

for number in numbers:
    if number > 0:
        positive += 1
    elif number < 0:
        negative += 1

print("Positive numbers:", positive)
print("Negative numbers:", negative)
````

## Find the Sum Using a Loop

````python
numbers = (10, 20, 30, 40, 50)

total = 0

for number in numbers:
    total += number

print("Total:", total)
````

## Find the Largest Element Using a Loop

````python
numbers = (25, 10, 45, 30, 60, 15)

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest:", largest)
````

## Find the Smallest Element Using a Loop

````python
numbers = (25, 10, 45, 30, 60, 15)

smallest = numbers[0]

for number in numbers:
    if number < smallest:
        smallest = number

print("Smallest:", smallest)
````

## 6. Loop Through Nested Tuples

Suppose we have student records:

```python
students = (
    ("Rahul", 85),
    ("Shilpa", 92),
    ("Amit", 78)
)

````python
students = (
    ("Rahul", 85),
    ("Shilpa", 92),
    ("Amit", 78)
)

for name, marks in students:
    print(name, marks)
````

## Find Students Who Scored Above 80

````python
students = (
    ("Rahul", 85),
    ("Shilpa", 92),
    ("Amit", 78),
    ("Priya", 88)
)

for name, marks in students:
    if marks > 80:
        print(name, marks)
````

## 7. Nested Loops with Tuples

A nested loop is a loop inside another loop.

This is useful when a tuple contains multiple tuples.

````python
data = (
    ("A", 10, 20),
    ("B", 30, 40),
    ("C", 50, 60)
)

for row in data:
    for value in row:
        print(value)
````

## Nested Loop with Conditions

````python
data = (
    (10, 20, 30),
    (40, 50, 60),
    (70, 80, 90)
)

for row in data:
    for value in row:
        if value > 50:
            print(value)
````

## 8. enumerate() with Nested Tuples

````python
students = (
    ("Rahul", 85),
    ("Shilpa", 92),
    ("Amit", 78)
)

for index, student in enumerate(students, start=1):
    print("Student", index, ":", student)
````

## 9. Sales Analysis Using a Tuple

Let's analyze monthly sales using a loop.

````python
sales = (
    ("January", 12000),
    ("February", 15000),
    ("March", 11000),
    ("April", 18000),
    ("May", 20000)
)

total = 0

for month, amount in sales:
    total += amount

print("Total Sales:", total)
````

````python
average = total / len(sales)

print("Average Sales:", average)
````

## Find Months Above Average

````python
for month, amount in sales:
    if amount > average:
        print(month, amount)
````

## Find the Month with Highest Sales

````python
highest_month = sales[0][0]
highest_sales = sales[0][1]

for month, amount in sales:
    if amount > highest_sales:
        highest_sales = amount
        highest_month = month

print("Highest Sales Month:", highest_month)
print("Sales:", highest_sales)
````
