## Program 1: Find the Sum of All Elements

Write a program to calculate the sum of all numeric elements in a tuple.

```python
numbers = (10, 20, 30, 40, 50)

total = 0

for number in numbers:
    total += number

print("Sum:", total)
```

## Program 2: Find the Average

Calculate the average of all elements in a numeric tuple.

```python
numbers = (10, 20, 30, 40, 50)

average = sum(numbers) / len(numbers)

print("Average:", average)
```

## Program 3: Find the Largest Element

Find the largest element without using `max()`.

```python
numbers = (25, 10, 45, 30, 60, 15)

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest:", largest)
```

## Program 4: Find the Smallest Element

Find the smallest element without using `min()`.

```python
numbers = (25, 10, 45, 30, 60, 15)

smallest = numbers[0]

for number in numbers:
    if number < smallest:
        smallest = number

print("Smallest:", smallest)
```

## Program 5: Count Occurrences of an Element

Find how many times a particular element appears in a tuple.

```python
numbers = (10, 20, 10, 30, 10, 40)

target = 10

count = numbers.count(target)

print(target, "appears", count, "times")
```

## Program 6: Search for an Element

Check whether a particular element exists in a tuple.

```python
numbers = (10, 20, 30, 40, 50)

target = 30

if target in numbers:
    print("Element found")
else:
    print("Element not found")
```

## Program 7: Find the Position of an Element

Find the index of the first occurrence of an element.

```python
numbers = (10, 20, 30, 40, 50)

target = 30

if target in numbers:
    print("Index:", numbers.index(target))
else:
    print("Element not found")
```

## Program 8: Separate Even and Odd Numbers

Print even and odd numbers separately.

```python
numbers = (10, 15, 20, 25, 30, 35, 40)

even = []
odd = []

for number in numbers:
    if number % 2 == 0:
        even.append(number)
    else:
        odd.append(number)

print("Even:", tuple(even))
print("Odd:", tuple(odd))
```

## Program 9: Count Even and Odd Numbers

```python
numbers = (10, 15, 20, 25, 30, 35, 40)

even_count = 0
odd_count = 0

for number in numbers:
    if number % 2 == 0:
        even_count += 1
    else:
        odd_count += 1

print("Even count:", even_count)
print("Odd count:", odd_count)
```

## Program 10: Reverse a Tuple

Reverse a tuple using slicing.

```python
numbers = (10, 20, 30, 40, 50)

reversed_tuple = numbers[::-1]

print(reversed_tuple)
```

## Program 11: Reverse Without Using Slicing

Reverse a tuple using a loop.

```python
numbers = (10, 20, 30, 40, 50)

reversed_list = []

for i in range(len(numbers) - 1, -1, -1):
    reversed_list.append(numbers[i])

reversed_tuple = tuple(reversed_list)

print(reversed_tuple)
```

## Program 12: Remove Duplicate Elements

Create a new tuple containing only unique elements.

Do not use `set()`.

```python
numbers = (10, 20, 10, 30, 20, 40, 10)

unique = []

for number in numbers:
    if number not in unique:
        unique.append(number)

result = tuple(unique)

print(result)
```

## Program 13: Merge Two Tuples

```python
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

merged = tuple1 + tuple2

print(merged)
```

## Program 14: Find Common Elements

Find elements that appear in both tuples.

```python
tuple1 = (10, 20, 30, 40, 50)
tuple2 = (30, 40, 50, 60, 70)

common = []

for number in tuple1:
    if number in tuple2:
        common.append(number)

print(tuple(common))
```

## Program 15: Find the Second Largest Element

Find the second largest unique value in a tuple.

```python
numbers = (10, 50, 20, 80, 30, 80, 60)

unique = []

for number in numbers:
    if number not in unique:
        unique.append(number)

unique.sort()

print("Second largest:", unique[-2])
```

## Program 16: Find the Second Smallest Element

```python
numbers = (40, 10, 30, 20, 10, 50)

unique = []

for number in numbers:
    if number not in unique:
        unique.append(number)

unique.sort()

print("Second smallest:", unique[1])
```

## Program 17: Find Frequency of Every Element

Count how many times every unique element appears.

```python
numbers = (10, 20, 10, 30, 20, 10, 40)

unique = []

for number in numbers:
    if number not in unique:
        unique.append(number)

for number in unique:
    print(number, "->", numbers.count(number))
```

## Program 18: Find the Most Frequently Occurring Element

```python
numbers = (10, 20, 10, 30, 20, 10, 40)

most_frequent = numbers[0]

for number in numbers:
    if numbers.count(number) > numbers.count(most_frequent):
        most_frequent = number

print("Most frequent:", most_frequent)
print("Frequency:", numbers.count(most_frequent))
```

## Program 19: Find the Least Frequently Occurring Element

```python
numbers = (10, 20, 10, 30, 20, 10, 40)

least_frequent = numbers[0]

for number in numbers:
    if numbers.count(number) < numbers.count(least_frequent):
        least_frequent = number

print("Least frequent:", least_frequent)
print("Frequency:", numbers.count(least_frequent))
```

## Program 20: Check Whether a Tuple is a Palindrome

A tuple is a palindrome if it reads the same forward and backward.

```python
numbers = (1, 2, 3, 2, 1)

if numbers == numbers[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

## Program 23: Find the Top Student

Use a nested tuple containing student names and marks.

```python
students = (
    ("Rahul", 85),
    ("Shilpa", 92),
    ("Amit", 78),
    ("Priya", 88)
)

top_student = students[0]

for student in students:
    if student[1] > top_student[1]:
        top_student = student

print("Top student:", top_student[0])
print("Marks:", top_student[1])
```

## Program 24: Sales Analysis

Find total, average, highest, and lowest sales.

```python
sales = (
    ("January", 12000),
    ("February", 15000),
    ("March", 11000),
    ("April", 18000),
    ("May", 20000)
)

amounts = []

for month, amount in sales:
    amounts.append(amount)

print("Total:", sum(amounts))
print("Average:", sum(amounts) / len(amounts))
print("Highest:", max(amounts))
print("Lowest:", min(amounts))
```

## Program 25: Find Months with Above-Average Sales

```python
sales = (
    ("January", 12000),
    ("February", 15000),
    ("March", 11000),
    ("April", 18000),
    ("May", 20000)
)

total = 0

for month, amount in sales:
    total += amount

average = total / len(sales)

print("Average sales:", average)
print("Months above average:")

for month, amount in sales:
    if amount > average:
        print(month, amount)
```
