# Introduction to Tuples

## Topic

Introduction to Tuples in Python

## Description

A tuple is an ordered collection of elements in Python.

Tuples are similar to lists, but tuples are immutable, which means their elements cannot be changed after creation.

**Author:** Shilpa

## What is a Tuple?

A tuple is a collection of elements enclosed in parentheses `()`.

Example:

```python
numbers = (10, 20, 30, 40)

````python

numbers = (10, 20, 30, 40)

print(numbers)
print(type(numbers))
````

## Creating a Tuple

We can create a tuple by placing multiple values inside parentheses.

Tuples can contain:

- Integers
- Floats
- Strings
- Boolean values
- Different data types

````python
student = ("Shilpa", 21, 85.5, True)

print(student)
````

## Tuple with Different Data Types

````python
data = ("Python", 100, 95.5, True)

print(data)
````

## Single-Element Tuple

A single-element tuple requires a comma.

Correct:

```python
numbers = (10,)

````python
a = (10,)

print(a)
print(type(a))
````

````python
b = (10)

print(b)
print(type(b))
````

## Tuple is Ordered

Tuples maintain the order in which elements are stored.

Example:

```python
fruits = ("Apple", "Banana", "Mango")

````python
fruits = ("Apple", "Banana", "Mango")

print(fruits)
````

## Tuple is Immutable

The most important difference between a list and a tuple is that a tuple is **immutable**.

Immutable means we cannot directly change, add, or remove elements after the tuple is created.

````python
numbers = (10, 20, 30)

print(numbers)

# This will cause an error:
# numbers[0] = 100
````

## List vs Tuple

| Feature | List | Tuple |
|---|---|---|
| Brackets | `[]` | `()` |
| Ordered | Yes | Yes |
| Mutable | Yes | No |
| Duplicates | Allowed | Allowed |
| Indexing | Yes | Yes |
| Slicing | Yes | Yes |

## When Should We Use Tuples?

Tuples are useful when data should not be changed.

Examples:

- Days of the week
- Months
- Coordinates
- Fixed configuration values
- Employee records
- Database records

````python
days = ("Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday")

print(days)
````

````python
coordinates = (18.5204, 73.8567)

print("Coordinates:", coordinates)
````

## Tuple with Duplicate Values

Tuples allow duplicate elements.

````python
numbers = (10, 20, 10, 30, 10)

print(numbers)
````

## Length of a Tuple

We can use the `len()` function to find the number of elements.

````python
fruits = ("Apple", "Banana", "Mango", "Orange")

print("Number of fruits:", len(fruits))
````
