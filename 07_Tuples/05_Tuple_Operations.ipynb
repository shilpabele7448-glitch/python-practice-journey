## 1. Tuple Concatenation

Concatenation means joining two or more tuples together.

We use the `+` operator.

### Syntax

```python
tuple1 + tuple2

````python
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

result = tuple1 + tuple2

print(result)
````

## Concatenating Tuples with Different Data Types

Tuples can contain different types of data, so we can also concatenate tuples containing different types.

````python
student_info = ("Shilpa", 21)
marks_info = (85, 90, 88)

result = student_info + marks_info

print(result)
````

## 2. Tuple Repetition

We can repeat the elements of a tuple using the `*` operator.

### Syntax

```python
tuple * number

````python
numbers = (10, 20, 30)

result = numbers * 3

print(result)
````

````python
fruits = ("Apple", "Mango")

print(fruits * 2)
````

## 3. Membership Operators

Membership operators check whether an element exists inside a tuple.

There are two membership operators:

- `in`
- `not in`

Example:

```python
10 in numbers

````python
numbers = (10, 20, 30, 40, 50)

print(30 in numbers)
print(100 in numbers)
````

````python
numbers = (10, 20, 30, 40, 50)

print(30 not in numbers)
print(100 not in numbers)
````

## Using Membership Operators with Conditions

````python
fruits = ("Apple", "Banana", "Mango", "Orange")

fruit = "Mango"

if fruit in fruits:
    print("Fruit is available")
else:
    print("Fruit is not available")
````

## 4. Comparing Tuples

Tuples can be compared using:

- `==`
- `!=`
- `<`
- `>`
- `<=`
- `>=`

Python compares tuples element by element.

````python
tuple1 = (10, 20, 30)
tuple2 = (10, 20, 30)

print(tuple1 == tuple2)
````

````python
tuple1 = (10, 20, 30)
tuple2 = (10, 20, 40)

print(tuple1 == tuple2)
print(tuple1 != tuple2)
````

## How Does Tuple Comparison Work?

Python compares the first elements.

If they are equal, it compares the second elements.

If those are equal, it continues to the next elements.

Example:

```python
(10, 20, 30)
(10, 20, 40)

````python
tuple1 = (10, 20, 30)
tuple2 = (10, 20, 40)

print(tuple1 < tuple2)
print(tuple1 > tuple2)
````

## 5. Finding the Length of a Tuple

The `len()` function returns the number of elements in a tuple.

````python
numbers = (10, 20, 30, 40, 50)

print(len(numbers))
````

## 6. Finding the Sum

The `sum()` function calculates the total of numeric elements.

````python
numbers = (10, 20, 30, 40, 50)

print(sum(numbers))
````

## 7. Finding the Maximum Value

The `max()` function returns the largest value.

````python
numbers = (10, 50, 20, 80, 30)

print(max(numbers))
````

## 8. Finding the Minimum Value

The `min()` function returns the smallest value.

````python
numbers = (10, 50, 20, 80, 30)

print(min(numbers))
````

## 9. Basic Tuple Analysis

We can combine different functions to analyze a tuple of numbers.

````python
marks = (78, 85, 92, 67, 88, 74)

print("Marks:", marks)
print("Number of subjects:", len(marks))
print("Total:", sum(marks))
print("Highest:", max(marks))
print("Lowest:", min(marks))
print("Average:", sum(marks) / len(marks))
````

## 10. Combining Concatenation and Repetition

````python
a = (1, 2)
b = (3, 4)

result = (a + b) * 2

print(result)
````

## Sales Data Example

Suppose we have monthly sales stored in a tuple.

We can use tuple operations to perform basic analysis.

````python
sales = (12000, 15000, 11000, 18000, 20000, 17000)

print("Sales:", sales)
print("Total Sales:", sum(sales))
print("Highest Sales:", max(sales))
print("Lowest Sales:", min(sales))
print("Average Sales:", sum(sales) / len(sales))
````

## Important Point

Some operations create a new tuple rather than modifying the existing tuple.

For example:

```python
numbers = (10, 20, 30)

numbers + (40, 50)

````python
numbers = (10, 20, 30)

new_numbers = numbers + (40, 50)

print("Original:", numbers)
print("New:", new_numbers)
````
