# Tuple Indexing

## Topic

Accessing Tuple Elements Using Indexing

## Description

Tuple indexing allows us to access individual elements from a tuple.

Python uses zero-based indexing, meaning the first element has index 0.

**Author:** Shilpa

## What is Indexing?

Indexing means accessing an individual element using its position.

Example:

```python
fruits = ("Apple", "Banana", "Mango", "Orange")

````python
fruits = ("Apple", "Banana", "Mango", "Orange")

print(fruits[0])
print(fruits[1])
print(fruits[2])
print(fruits[3])
````

## Using enumerate()

`enumerate()` gives us both the index and the value.

````python
fruits = ("Apple", "Banana", "Mango")

for index, fruit in enumerate(fruits):
    print(index, fruit)
````
