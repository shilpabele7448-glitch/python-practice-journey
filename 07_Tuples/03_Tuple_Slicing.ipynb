## Description

Tuple slicing allows us to extract a portion of a tuple using a start index, stop index, and optional step.

## What is Tuple Slicing?

Slicing is used to extract multiple elements from a tuple.

### Syntax

```python
tuple[start:stop]

````python
numbers = (10, 20, 30, 40, 50)

print(numbers[1:4])
````
