## What is a Nested Tuple?

A nested tuple is a tuple inside another tuple.

Example:

```python
students = (
    ("Rahul", 85),
    ("Shilpa", 92),
    ("Amit", 78)
)

````python
students = (
    ("Rahul", 85),
    ("Shilpa", 92),
    ("Amit", 78)
)

print(students)
print(type(students))
````

## Understanding the Structure

Our tuple looks like:

```text
students
   |
   ├── ("Rahul", 85)
   ├── ("Shilpa", 92)
   └── ("Amit", 78)

````python
students = (
    ("Rahul", 85),
    ("Shilpa", 92),
    ("Amit", 78)
)

print(students[0])
print(students[1])
print(students[2])
````

## Accessing Elements from an Inner Tuple

We can use two indexes:

```python
outer_tuple[index1][index2]

````python
students = (
    ("Rahul", 85),
    ("Shilpa", 92),
    ("Amit", 78)
)

print(students[0][0])
print(students[0][1])
````

````python
print(students[1][0])
print(students[1][1])

print(students[2][0])
print(students[2][1])
````

## Accessing Specific Information

We can directly access a particular student's name or marks.

For example:

```python
students[1][0]

````python
print("Second student's name:", students[1][0])
print("Second student's marks:", students[1][1])
````

## Loop Through Inner Tuples

We can use a `for` loop to process every inner tuple.

````python
students = (
    ("Rahul", 85),
    ("Shilpa", 92),
    ("Amit", 78)
)

for student in students:
    print(student)
````

## Unpacking Nested Tuples

Instead of storing the entire inner tuple in one variable, we can unpack it into separate variables.

````python
for name, marks in students:
    print("Name:", name)
    print("Marks:", marks)
    print()
````

## Finding Total Marks

````python
total = 0

for name, marks in students:
    total += marks

print("Total Marks:", total)
````

## Finding Average Marks

````python
average = total / len(students)

print("Average Marks:", average)
````

## Finding Highest Marks

````python
highest = students[0][1]

for name, marks in students:
    if marks > highest:
        highest = marks

print("Highest Marks:", highest)
````

## Finding the Student with Highest Marks

````python
highest_name = students[0][0]
highest_marks = students[0][1]

for name, marks in students:
    if marks > highest_marks:
        highest_marks = marks
        highest_name = name

print("Top Student:", highest_name)
print("Marks:", highest_marks)
````

## Nested Tuple with Employee Records

Nested tuples can be used to represent simple structured records.

Each employee will have:

- Employee ID
- Name
- Department
- Salary

````python
employees = (
    (101, "Rahul", "IT", 50000),
    (102, "Shilpa", "HR", 45000),
    (103, "Amit", "Finance", 60000),
    (104, "Priya", "IT", 55000)
)

print(employees)
````

````python
for emp_id, name, department, salary in employees:
    print("ID:", emp_id)
    print("Name:", name)
    print("Department:", department)
    print("Salary:", salary)
    print()
````

## Find Employees from IT Department

````python
for emp_id, name, department, salary in employees:
    if department == "IT":
        print(name)
````

## Find Highest Salary

````python
highest_salary = employees[0][3]

for emp_id, name, department, salary in employees:
    if salary > highest_salary:
        highest_salary = salary

print("Highest Salary:", highest_salary)
````

## Find Employee with Highest Salary

````python
highest_employee = employees[0]

for employee in employees:
    if employee[3] > highest_employee[3]:
        highest_employee = employee

print("Employee ID:", highest_employee[0])
print("Name:", highest_employee[1])
print("Department:", highest_employee[2])
print("Salary:", highest_employee[3])
````

## Slicing a Nested Tuple

We can slice the outer tuple just like a normal tuple.

````python
students = (
    ("Rahul", 85),
    ("Shilpa", 92),
    ("Amit", 78),
    ("Priya", 88)
)

print(students[:2])
````

## Important: Nested Tuples and Immutability

Nested tuples are also immutable.

We cannot directly change an inner tuple's element.

For example:

```python
students[0][1] = 95


````python
# Can get a TypeError.
students = (
    ("Rahul", 85),
    ("Shilpa", 92)
)

students[0][1] = 95
````
