# Tuple Methods

## Topic

Built-in Methods of Tuples in Python

## Description

Tuples provide two main built-in methods:

- `count()`
- `index()`

These methods allow us to count occurrences of an element and find the position of an element.

**Author:** Shilpa

## 1. count()

The `count()` method returns the number of times a particular element appears in a tuple.

### Syntax

```python
tuple.count(element)

````python
numbers = (10, 20, 10, 30, 10, 40)

result = numbers.count(10)

print(result)
````

## Counting Different Elements

````python
fruits = ("Apple", "Banana", "Apple", "Mango", "Apple")

print("Apple:", fruits.count("Apple"))
print("Banana:", fruits.count("Banana"))
print("Mango:", fruits.count("Mango"))
````

## Counting an Element That Does Not Exist

If the element is not present, `count()` returns `0`.

````python
numbers = (10, 20, 30, 40)

print(numbers.count(50))
````

## 2. index()

The `index()` method returns the position of the first occurrence of an element.

### Syntax

```python
tuple.index(element)

````python
numbers = (10, 20, 30, 40)

position = numbers.index(30)

print(position)
````

## index() with Duplicate Values

If an element appears multiple times, `index()` returns the index of its **first occurrence**.

````python
numbers = (10, 20, 10, 30, 10)

print(numbers.index(10))
````

## Element Not Found

If we use `index()` for an element that does not exist, Python raises a `ValueError`.

````python
numbers = (10, 20, 30)

print(numbers.index(50))
````

## Using index() Safely

We can check whether an element exists before using `index()`.

````python
numbers = (10, 20, 30, 40)

target = 30

if target in numbers:
    print("Index:", numbers.index(target))
else:
    print("Element not found")
````

## count() vs index()

| Method | Purpose |
|---|---|
| `count()` | Counts occurrences |
| `index()` | Finds first position |

## Finding the Position of Different Elements

````python
fruits = ("Apple", "Banana", "Mango", "Orange")

for fruit in fruits:
    print(fruit, "->", fruits.index(fruit))
````

## Counting Values in a Tuple

Let's count how many times each unique value appears.

````python
numbers = (10, 20, 10, 30, 20, 10)

unique_numbers = []

for number in numbers:
    if number not in unique_numbers:
        unique_numbers.append(number)

for number in unique_numbers:
    print(number, "->", numbers.count(number))
````
