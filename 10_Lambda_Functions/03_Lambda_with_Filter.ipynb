# 🐍 Lambda Functions with filter()

The `filter()` function is used to select items from an iterable
based on a condition.

In this notebook, we will learn:

1. What is filter()?
2. Basic filter()
3. filter() with Lambda
4. Filtering numbers
5. Filtering even and odd numbers
6. Filtering positive and negative numbers
7. Filtering strings
8. Filtering salaries
9. Filtering sales data
10. Data Analytics examples
11. Practice questions
12. Mini challenge

```python
numbers = [10, 15, 20, 25, 30]

result = []

for number in numbers:
    if number > 20:
        result.append(number)

print(result)
```

```python
numbers = [10, 15, 20, 25, 30]

def greater_than_20(number):
    return number > 20

result = filter(greater_than_20, numbers)

print(list(result))
```

```python
numbers = [10, 15, 20, 25, 30]

result = list(filter(lambda x: x > 20, numbers))

print(result)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even_numbers = list(
    filter(lambda x: x % 2 == 0, numbers)
)

print(even_numbers)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

odd_numbers = list(
    filter(lambda x: x % 2 != 0, numbers)
)

print(odd_numbers)
```

```python
numbers = [-10, -5, 0, 5, 10, 15]

positive_numbers = list(
    filter(lambda x: x > 0, numbers)
)

print(positive_numbers)
```

```python
numbers = [-10, -5, 0, 5, 10, 15]

negative_numbers = list(
    filter(lambda x: x < 0, numbers)
)

print(negative_numbers)
```

```python
numbers = [20, 45, 60, 75, 30, 90]

result = list(
    filter(lambda x: x > 50, numbers)
)

print(result)
```

```python
numbers = [10, 20, 25, 30, 45, 50, 60]

result = list(
    filter(lambda x: 20 <= x <= 50, numbers)
)

print(result)
```

```python
names = ["Amit", "Priya", "Rahul", "Raj", "Sneha"]

result = list(
    filter(lambda name: len(name) > 4, names)
)

print(result)
```

```python
names = ["Amit", "Priya", "Anjali", "Rahul", "Asha"]

result = list(
    filter(lambda name: name.startswith("A"), names)
)

print(result)
```

```python
salaries = [22000, 28000, 35000, 45000, 60000]

high_salaries = list(
    filter(lambda salary: salary >= 40000, salaries)
)

print(high_salaries)
```

```python
sales = [5000, 12000, 18000, 7000, 25000, 30000]

high_sales = list(
    filter(lambda sale: sale >= 15000, sales)
)

print(high_sales)
```

```python
marks = [35, 67, 42, 28, 75, 90, 31]

passed = list(
    filter(lambda mark: mark >= 40, marks)
)

print(passed)
```

```python
customers = [
    {"name": "Amit", "purchase": 5000},
    {"name": "Priya", "purchase": 15000},
    {"name": "Rahul", "purchase": 8000},
    {"name": "Sneha", "purchase": 20000}
]

high_value_customers = list(
    filter(lambda customer: customer["purchase"] >= 10000, customers)
)

print(high_value_customers)
```

```python
sales = [5000, 10000, 12000, 15000, 18000, 21000]

result = list(
    filter(lambda sale: sale > 10000 and sale % 2 == 0, sales)
)

print(result)
```

# 🧠 map() vs filter()

## map()

Used when we want to TRANSFORM every item.

Example:

numbers = [1, 2, 3, 4]

list(map(lambda x: x * 2, numbers))

Result:

[2, 4, 6, 8]


## filter()

Used when we want to SELECT specific items.

Example:

numbers = [1, 2, 3, 4]

list(filter(lambda x: x > 2, numbers))

Result:

[3, 4]

```python
salaries = [22000, 30000, 35000, 45000, 50000]

high_salaries = filter(
    lambda salary: salary >= 30000,
    salaries
)

new_salaries = list(
    map(lambda salary: salary * 1.10, high_salaries)
)

print(new_salaries)
```

```python
salaries = [22000, 30000, 35000, 45000, 50000]

high_salaries = filter(
    lambda salary: salary >= 30000,
    salaries
)

new_salaries = list(
    map(lambda salary: round(salary * 1.10, 2),
        high_salaries)
)

print(new_salaries)
```
