# 🐍 Lambda Functions with map()

The `map()` function is used to apply a function to every item
in an iterable such as a list.

In this notebook, we will learn:

1. What is map()?
2. Basic map()
3. map() with lambda
4. Converting map output to a list
5. Mathematical calculations
6. Working with strings
7. Working with multiple lists
8. Data Analytics examples
9. Practice questions
10. Mini challenge
11. 

```python
numbers = [1, 2, 3, 4, 5]

result = []

for number in numbers:
    result.append(number * 2)

print(result)
```

```python
numbers = [1, 2, 3, 4, 5]

def double(number):
    return number * 2

result = map(double, numbers)

print(list(result))
```

```python
numbers = [1, 2, 3, 4, 5]

result = map(lambda x: x * 2, numbers)

print(list(result))
```

```python
numbers = [10, 20, 30]

result = map(lambda x: x + 5, numbers)

print(result)
```

```python
numbers = [10, 20, 30]

result = list(map(lambda x: x + 5, numbers))

print(result)
```

```python
numbers = [2, 4, 6, 8]

squares = list(map(lambda x: x ** 2, numbers))

print(squares)
```

```python
numbers = [1, 2, 3, 4]

cubes = list(map(lambda x: x ** 3, numbers))

print(cubes)
```

```python
numbers = [5, 10, 15, 20]

result = list(map(lambda x: x + 10, numbers))

print(result)
```

```python
prices = [100, 200, 500, 1000]

gst_prices = list(map(lambda price: price * 1.18, prices))

print(gst_prices)
```

```python
prices = [1000, 2000, 3000, 4000]

discounted_prices = list(
    map(lambda price: price * 0.90, prices)
)

print(discounted_prices)
```

```python
names = ["amit", "priya", "rahul", "sneha"]

uppercase_names = list(map(lambda name: name.upper(), names))

print(uppercase_names)
```

```python
names = ["Amit", "Priya", "Rahul", "Sneha"]

lengths = list(map(lambda name: len(name), names))

print(lengths)
```

```python
values = ["10", "20", "30", "40"]

numbers = list(map(lambda x: int(x), values))

print(numbers)
```

```python
prices = [100, 200, 300]
quantities = [2, 3, 4]

total = list(
    map(lambda price, quantity: price * quantity,
        prices,
        quantities)
)

print(total)
```

```python
salaries = [25000, 30000, 40000, 50000]

new_salaries = list(
    map(lambda salary: salary * 1.10, salaries)
)

print(new_salaries)
```

```python
sales = [10000, 15000, 20000, 25000]

sales_with_tax = list(
    map(lambda sale: sale * 1.18, sales)
)

print(sales_with_tax)
```

## map() vs for Loop

Using a loop:

result = []

for number in numbers:
    result.append(number * 2)

Using map():

result = list(map(lambda x: x * 2, numbers))

Both can produce the same result.

`map()` is especially useful when applying the same simple
transformation to every item.

```python
numbers = [1, 2, 3]

result = map(lambda x: x * 2, numbers)

print(result)
```
