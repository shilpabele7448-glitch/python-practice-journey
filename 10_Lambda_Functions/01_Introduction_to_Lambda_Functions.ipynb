# 🐍 Introduction to Lambda Functions

Lambda functions are small, anonymous functions in Python.

They are useful when we need a simple function for a short operation.

In this notebook, we will learn:

1. What is a lambda function?
2. Lambda syntax
3. Lambda with one argument
4. Lambda with multiple arguments
5. Lambda with numbers
6. Lambda with strings
7. Lambda with conditions
8. Lambda vs normal functions
9. Practical Data Analytics examples
10. Practice questions

```python
def square(x):
    return x * x

print(square(5))
```

```python
square = lambda x: x * x

print(square(5))
```

## Lambda Syntax

lambda arguments: expression

Example:

lambda x: x * 2

```python
double = lambda x: x * 2

print(double(10))
```

```python
cube = lambda x: x ** 3

print(cube(4))
```

```python
add = lambda a, b: a + b

print(add(10, 20))
```

```python
multiply = lambda a, b: a * b

print(multiply(5, 4))
```

```python
total = lambda a, b, c: a + b + c

print(total(10, 20, 30))
```

```python
length = lambda text: len(text)

print(length("Python"))
```

```python
uppercase = lambda text: text.upper()

print(uppercase("data analytics"))
```

```python
check = lambda x: "Even" if x % 2 == 0 else "Odd"

print(check(10))
print(check(7))
```

```python
check_number = lambda x: "Positive" if x > 0 else "Negative"

print(check_number(10))
print(check_number(-5))
```

```python
check_number = lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"

print(check_number(10))
print(check_number(-5))
print(check_number(0))
```

## Lambda vs Normal Function

Normal Function:

def square(x):
    return x * x

Lambda Function:

square = lambda x: x * x

### Normal Function
- Can contain multiple statements
- Better for complex logic
- Easier to read for larger programs
- Has a function name

### Lambda Function
- Usually one expression
- Short and simple
- Useful for temporary operations
- Anonymous function

```python
result = lambda x: x + 10

print(result(5))
```

```python
prices = [100, 200, 300, 400]

add_tax = lambda price: price * 1.18

print(add_tax(100))
```

```python
salaries = [25000, 35000, 45000, 55000]

increase = lambda salary: salary * 1.10

print(increase(25000))
```

```python
employee = {
    "name": "Rahul",
    "salary": 45000
}

get_salary = lambda data: data["salary"]

print(get_salary(employee))
```

```python
names = ["Amit", "Priya", "Raj", "Sneha"]

get_length = lambda name: len(name)

for name in names:
    print(name, get_length(name))
```
