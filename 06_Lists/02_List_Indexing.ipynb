## Topic

List Indexing

## Description

List indexing is used to access individual elements from a list.

Python uses zero-based indexing, which means the first element has index 0.

**Author:** Shilpa

## What is List Indexing?

Every item in a list has a position called an **index**.

Python starts counting from **0**.

Example:

```text
Index :   0        1        2
List  : ["Apple","Banana","Mango"]
```

````python
fruits = ["Apple", "Banana", "Mango"]

print(fruits[0])
print(fruits[1])
print(fruits[2])
````

## Negative Indexing

Negative indexing starts from the end of the list.

```text
Index :  -3       -2       -1
List  : ["Apple","Banana","Mango"]
```

````python
fruits = ["Apple", "Banana", "Mango"]

print(fruits[-1])
print(fruits[-2])
print(fruits[-3])
````

## Accessing Different Data Types

````python
student = ["Shilpa", 22, "Pune", 89.5]

print(student[0])
print(student[1])
print(student[2])
print(student[3])
````

## Using Variables as Index

````python
numbers = [10, 20, 30, 40, 50]

index = 3

print(numbers[index])
````

## Finding the First and Last Element

````python
colors = ["Red", "Blue", "Green", "Yellow"]

print("First:", colors[0])
print("Last :", colors[-1])
````

## Finding the Middle Element

````python
numbers = [10, 20, 30, 40, 50]

middle = len(numbers) // 2

print(numbers[middle])
````
