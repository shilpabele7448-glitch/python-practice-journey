A list is one of the most commonly used data structures in Python. It allows us to store multiple values in a single variable.

## What is a List?

A list is a collection of items stored in a single variable.

Instead of creating many variables, we can store all related values in one list.

Lists are written using square brackets `[]`.

````python
fruits = ["Apple", "Banana", "Mango"]

print(fruits)
````

## Why Do We Use Lists?

Without a list:

```python
fruit1 = "Apple"
fruit2 = "Banana"
fruit3 = "Mango"
```

With a list:

```python
fruits = ["Apple", "Banana", "Mango"]
```

Using a list keeps the code shorter, cleaner, and easier to manage.

````python
numbers = [10, 20, 30, 40, 50]

print(numbers)
````

````python
mixed = ["Shilpa", 22, 85.5, True]

print(mixed)
````

## Characteristics of Lists

- Ordered
- Mutable (can be changed)
- Allows duplicate values
- Can store different data types
- Indexed (starts from 0)

````python
colors = ["Red", "Blue", "Green", "Blue"]

print(colors)
````

````python
empty_list = []

print(empty_list)
````

````python
marks = [78, 85, 90]

print(type(marks))
````
