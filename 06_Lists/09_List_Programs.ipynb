## Program 1: Find the Sum of a List

Write a program to calculate the sum of all elements in a list.

````python
numbers = [10, 20, 30, 40, 50]

total = 0

for number in numbers:
    total += number

print("Sum:", total)
````

## Program 2: Find the Average of a List

````python
numbers = [10, 20, 30, 40, 50]

total = 0

for number in numbers:
    total += number

average = total / len(numbers)

print("Average:", average)
````

## Program 3: Find the Largest Element

Find the largest number without using the built-in `max()` function.

````python
numbers = [25, 10, 45, 30, 15]

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest:", largest)
````

## Program 4: Find the Smallest Element

Find the smallest number without using the built-in `min()` function.

````python
numbers = [25, 10, 45, 30, 15]

smallest = numbers[0]

for number in numbers:
    if number < smallest:
        smallest = number

print("Smallest:", smallest)
````

## Program 5: Count Occurrences

Count how many times a particular element appears in a list.

````python
numbers = [10, 20, 10, 30, 10, 40]

target = 10
count = 0

for number in numbers:
    if number == target:
        count += 1

print("Count:", count)
````

## Program 6: Find Common Elements

Find the elements that are present in both lists.

````python
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]

common = []

for number in list1:
    if number in list2:
        common.append(number)

print("Common elements:", common)
````

## Program 7: Merge Two Lists

````python
list1 = [1, 2, 3]
list2 = [4, 5, 6]

merged = list1 + list2

print("Merged list:", merged)
````

## Program 8: Remove Duplicates

Create a new list containing only unique elements.

Do not use `set()`.

````python
numbers = [1, 2, 2, 3, 4, 4, 5]

unique = []

for number in numbers:
    if number not in unique:
        unique.append(number)

print("Unique elements:", unique)
````

## Program 9: Reverse a List

Reverse a list without using the `reverse()` method.

````python
numbers = [10, 20, 30, 40, 50]

reversed_list = []

for number in numbers:
    reversed_list.insert(0, number)

print("Reversed list:", reversed_list)
````

## Program 10: Find the Second Largest Number

````python
numbers = [10, 40, 20, 50, 30]

unique_numbers = []

for number in numbers:
    if number not in unique_numbers:
        unique_numbers.append(number)

unique_numbers.sort()

print("Second largest:", unique_numbers[-2])
````

## Program 11: Find the Second Smallest Number

````python
numbers = [10, 40, 20, 50, 30]

unique_numbers = []

for number in numbers:
    if number not in unique_numbers:
        unique_numbers.append(number)

unique_numbers.sort()

print("Second smallest:", unique_numbers[1])
````

## Program 12: Separate Even and Odd Numbers

````python
numbers = [10, 15, 22, 31, 40, 55]

even = []
odd = []

for number in numbers:
    if number % 2 == 0:
        even.append(number)
    else:
        odd.append(number)

print("Even:", even)
print("Odd:", odd)
````

## Program 13: Find All Numbers Greater Than the Average

````python
numbers = [10, 20, 30, 40, 50]

total = 0

for number in numbers:
    total += number

average = total / len(numbers)

greater_than_average = []

for number in numbers:
    if number > average:
        greater_than_average.append(number)

print("Average:", average)
print("Numbers greater than average:", greater_than_average)
````

## Program 14: Find the Missing Number

A list contains numbers from 1 to 5, but one number is missing.

Find the missing number.

````python
numbers = [1, 2, 3, 5]

for number in range(1, 6):
    if number not in numbers:
        print("Missing number:", number)
````

## Program 15: Remove All Occurrences of an Element

````python
numbers = [10, 20, 10, 30, 10, 40]

target = 10

result = []

for number in numbers:
    if number != target:
        result.append(number)

print("Result:", result)
````

## Program 16: Rotate a List to the Right

Move the last element to the beginning.

Example:

```text
Original: [1, 2, 3, 4, 5]

Result:   [5, 1, 2, 3, 4]

````python
numbers = [1, 2, 3, 4, 5]

rotated = numbers[-1:] + numbers[:-1]

print("Rotated list:", rotated)
````

## Program 17: Find the Frequency of Each Element

Count how many times every element appears in a list.

````python
numbers = [1, 2, 2, 3, 1, 2, 4]

frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number] += 1
    else:
        frequency[number] = 1

print(frequency)
````

## Program 18: Find the Common Elements Without Duplicates

````python
list1 = [1, 2, 2, 3, 4]
list2 = [2, 3, 3, 5, 6]

common = []

for number in list1:
    if number in list2 and number not in common:
        common.append(number)

print("Common elements:", common)
````

## Program 19: Find the Largest and Smallest Together

````python
numbers = [45, 12, 78, 23, 9, 56]

largest = numbers[0]
smallest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

    if number < smallest:
        smallest = number

print("Largest:", largest)
print("Smallest:", smallest)
````

## Program 20: Check Whether a List is Sorted

Check whether the elements are arranged in ascending order.

````python
numbers = [10, 20, 30, 40, 50]

is_sorted = True

for i in range(len(numbers) - 1):
    if numbers[i] > numbers[i + 1]:
        is_sorted = False
        break

if is_sorted:
    print("List is sorted")
else:
    print("List is not sorted")
````
