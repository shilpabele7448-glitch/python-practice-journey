It is commonly used for transforming and filtering data.

## What is List Comprehension?

List comprehension is a shorter way of creating a list using a loop.

Instead of writing:

```python
numbers = [1, 2, 3, 4, 5]

squares = []

for number in numbers:
    squares.append(number ** 2)

print(squares)
```

We can write:

```python
squares = [number ** 2 for number in numbers]

print(squares)
```

Both produce the same result.

````python
numbers = [1, 2, 3, 4, 5]

squares = [number ** 2 for number in numbers]

print(squares)
````

## Basic Syntax

```python
new_list = [expression for item in iterable]
```

Here:

- `expression` → What we want to put into the new list
- `item` → Each element
- `iterable` → Existing list or another iterable

````python
numbers = [1, 2, 3, 4, 5]

doubled = [number * 2 for number in numbers]

print(doubled)
````

## Converting Strings

List comprehension can also be used with strings.

````python
names = ["rahul", "shilpa", "amit"]

uppercase_names = [name.upper() for name in names]

print(uppercase_names)
````

## List Comprehension with if Condition

We can filter elements using an `if` condition.

### Syntax

```python
new_list = [expression for item in iterable if condition]
```

````python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = [number for number in numbers if number % 2 == 0]

print(even_numbers)
````

````python
numbers = [1, 2, 3, 4, 5, 6]

odd_numbers = [number for number in numbers if number % 2 != 0]

print(odd_numbers)
````

## Filtering Values Greater Than a Number

````python
numbers = [10, 25, 5, 40, 15, 30]

large_numbers = [number for number in numbers if number > 20]

print(large_numbers)
````

## List Comprehension with if-else

We can also use `if-else`.

### Syntax

```python
new_list = [value_if_true if condition else value_if_false for item in iterable]
```

````python
numbers = [1, 2, 3, 4, 5]

result = ["Even" if number % 2 == 0 else "Odd" for number in numbers]

print(result)
````

## Converting Marks into Pass or Fail

List comprehension can be useful for simple data transformations.

````python
marks = [85, 32, 67, 25, 90]

result = ["Pass" if mark >= 35 else "Fail" for mark in marks]

print(result)
````

## Working with Strings

We can filter words based on their length.

````python
words = ["Python", "AI", "Data", "Analytics", "SQL"]

long_words = [word for word in words if len(word) > 4]

print(long_words)
````

## Nested List Comprehension

List comprehension can also work with nested lists.

Example:

```python
matrix = [
    [1, 2],
    [3, 4],
    [5, 6]
]
```

````python
matrix = [
    [1, 2],
    [3, 4],
    [5, 6]
]

flattened = [number for row in matrix for number in row]

print(flattened)
````

## Normal Loop vs List Comprehension

### Normal Loop

```python
numbers = [1, 2, 3, 4, 5]

squares = []

for number in numbers:
    squares.append(number ** 2)
```

### List Comprehension

```python
squares = [number ** 2 for number in numbers]
```

List comprehension is more compact, but a normal loop can sometimes be easier to understand when the logic becomes complicated.
