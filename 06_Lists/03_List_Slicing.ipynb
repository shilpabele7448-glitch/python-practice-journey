# List Slicing

## Topic

List Slicing

## Description

List slicing is used to access multiple elements from a list. It allows us to extract a portion of a list using the slicing operator `:`.

**Author:** Shilpa

## What is List Slicing?

List slicing extracts a part of a list.

**Syntax:**

```python
list_name[start : stop : step]
```

- **start** → Starting index (included)
- **stop** → Ending index (excluded)
- **step** → Number of positions to skip

````python
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]

print(fruits[1:4])
````

## Slicing from the Beginning

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[:3])
````

## Slicing Until the End

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[2:])
````

## Copying a List

````python
numbers = [10, 20, 30, 40, 50]

copy_list = numbers[:]

print(copy_list)
````

## Using Step

````python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[::2])
````

## Reverse a List

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[::-1])
````

## Negative Index Slicing

````python
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]

print(fruits[-4:-1])
````

## Real-Life Examples

````python
marks = [78, 82, 90, 67, 88, 95]

print("First Three Marks :", marks[:3])
print("Last Three Marks  :", marks[-3:])
print("Alternate Marks   :", marks[::2])
````

## Data Analytics Connection

List slicing is useful when working with datasets.

Examples:

- Selecting the first 100 records
- Extracting specific columns
- Splitting training and testing data
- Creating subsets for analysis

Understanding slicing helps before learning NumPy and Pandas.

## Summary

- `start` is included.
- `stop` is excluded.
- `step` decides how many elements to skip.
- `[::-1]` reverses a list.
- `[:]` creates a shallow copy of the list.

# Conclusion

List slicing is one of the most useful features in Python. It makes it easy to extract, copy, and reverse data efficiently, and it is widely used in data analysis and machine learning.
