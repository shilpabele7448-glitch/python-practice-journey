## Description

Python provides built-in methods to add, insert, and remove elements from a list. These methods make working with lists easier and more efficient.

## What are List Methods?

List methods are built-in functions that perform operations on lists.

Some commonly used methods are:

- append()
- extend()
- insert()
- remove()
- pop()

## 1. append()

The `append()` method adds one element to the **end** of the list.

````python
fruits = ["Apple", "Banana"]

fruits.append("Mango")

print(fruits)
````

## 2. extend()

The `extend()` method adds multiple elements from another list.

````python
fruits = ["Apple", "Banana"]

fruits.extend(["Mango", "Orange"])

print(fruits)
````

## append() vs extend()

- append() adds **one object**.
- extend() adds **each element** of another iterable.

````python
numbers = [1, 2]

numbers.append([3, 4])

print(numbers)
````

````python
numbers = [1, 2]

numbers.extend([3, 4])

print(numbers)
````

## 3. insert()

The `insert()` method inserts an element at a specific index.

**Syntax**

```python
list.insert(index, value)
```

````python
fruits = ["Apple", "Banana", "Orange"]

fruits.insert(2, "Mango")

print(fruits)
````

## 4. remove()

The `remove()` method removes the first matching value from the list.

````python
fruits = ["Apple", "Banana", "Mango"]

fruits.remove("Banana")

print(fruits)
````

## 5. pop()

The `pop()` method removes an element using its index.

If no index is given, it removes the last element.

````python
numbers = [10, 20, 30, 40]

numbers.pop()

print(numbers)
````

````python
numbers = [10, 20, 30, 40]

numbers.pop(1)

print(numbers)
````

## Real-Life Example

````python
cart = ["Laptop", "Mouse"]

cart.append("Keyboard")

cart.remove("Mouse")

print(cart)
````

## Data Analytics Connection

## Summary

| Method | Purpose |
|---------|---------|
