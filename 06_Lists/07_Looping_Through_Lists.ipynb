# Looping Through Lists

## Topic

Looping Through Lists

## Description

Loops allow us to process each element of a list one by one.

In this notebook, we will learn how to use `for` loops, `while` loops, `enumerate()`, and other techniques to work with list data.

## Why Do We Use Loops with Lists?

A list may contain many values.

Instead of writing separate code for every element, we can use a loop to process all elements automatically.

Example:

```python
fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
    print(fruit)
```

````python
fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
    print(fruit)
````

## Understanding the Loop

```python
for fruit in fruits:
    print(fruit)
```

Here:

- `fruits` → list
- `fruit` → temporary variable
- The loop takes one element at a time.
- `print(fruit)` prints the current element.

````python
numbers = [10, 20, 30, 40, 50]

for number in numbers:
    print(number)
````

## Performing Calculations

We can perform calculations on each list element using a loop.

````python
numbers = [10, 20, 30, 40, 50]

for number in numbers:
    print(number * 2)
````

## Finding the Sum of a List

We can use a loop to calculate the total of all numbers.

````python
numbers = [10, 20, 30, 40, 50]

total = 0

for number in numbers:
    total = total + number

print("Total:", total)
````

## Finding the Average

Average can be calculated using:

```text
Average = Total / Number of Values
```

````python
numbers = [10, 20, 30, 40, 50]

total = 0

for number in numbers:
    total += number

average = total / len(numbers)

print("Average:", average)
````

## Finding Even Numbers

````python
numbers = [10, 15, 22, 31, 40, 55]

for number in numbers:
    if number % 2 == 0:
        print(number)
````

## Finding Odd Numbers

````python
numbers = [10, 15, 22, 31, 40, 55]

for number in numbers:
    if number % 2 != 0:
        print(number)
````

## Finding the Largest Number

We can compare the elements of a list to find the largest value.

````python
numbers = [25, 10, 45, 30, 15]

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest:", largest)
````

## Finding the Smallest Number

````python
numbers = [25, 10, 45, 30, 15]

smallest = numbers[0]

for number in numbers:
    if number < smallest:
        smallest = number

print("Smallest:", smallest)
````

## Searching for an Element

We can use the `in` operator to check whether an element exists in a list.

````python
fruits = ["Apple", "Banana", "Mango", "Orange"]

search = input("Enter fruit name: ")

if search in fruits:
    print("Fruit Found")
else:
    print("Fruit Not Found")
````

## Looping with Index

Sometimes we need both the **index** and the **value**.

We can use:

```python
range(len(list))
```

````python
fruits = ["Apple", "Banana", "Mango"]

for i in range(len(fruits)):
    print(i, fruits[i])
````

## Using enumerate()

`enumerate()` gives us both the index and the value.

It is usually cleaner than using `range(len(list))`.

````python
fruits = ["Apple", "Banana", "Mango"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
````

## Starting enumerate() from 1

We can change the starting number using `start=1`.

````python
fruits = ["Apple", "Banana", "Mango"]

for index, fruit in enumerate(fruits, start=1):
    print(index, fruit)
````

## Looping Through a Nested List

Nested lists can also be processed using loops.

````python
students = [
    ["Rahul", 85],
    ["Shilpa", 92],
    ["Amit", 78]
]

for student in students:
    print(student[0], "-", student[1])
````

## Using a While Loop

A `while` loop can also be used to process list elements.

However, a `for` loop is generally simpler when we want to visit every element of a list.

````python
numbers = [10, 20, 30, 40]

i = 0

while i < len(numbers):
    print(numbers[i])
    i += 1
````

## Real-World Example: Student Marks

````python
marks = [78, 85, 92, 67, 88]

total = 0

for mark in marks:
    total += mark

average = total / len(marks)

print("Total Marks:", total)
print("Average Marks:", average)
````
