# Nested Lists

## Topic

Nested Lists

## Description

A nested list is a list that contains one or more lists as its elements.

Nested lists are useful for representing tables, matrices, student records, and other structured data.

**Author:** Shilpa

## What is a Nested List?

A nested list is simply a **list inside another list**.

Example:

```python
students = [
    ["Rahul", 85],
    ["Shilpa", 92],
    ["Amit", 78]
]
```

Each inner list stores information about one student.

````python
students = [
    ["Rahul", 85],
    ["Shilpa", 92],
    ["Amit", 78]
]

print(students)
````

## Accessing Elements

Use two indexes:

```python
list[row][column]
```

- First index → Row
- Second index → Column

````python
students = [
    ["Rahul", 85],
    ["Shilpa", 92],
    ["Amit", 78]
]

print(students[0][0])
print(students[0][1])
````

## Accessing Another Row

````python
students = [
    ["Rahul", 85],
    ["Shilpa", 92],
    ["Amit", 78]
]

print(students[1][0])
print(students[1][1])
````

## Modifying Nested Lists

````python
students = [
    ["Rahul", 85],
    ["Shilpa", 92],
    ["Amit", 78]
]

students[2][1] = 80

print(students)
````

## Loop Through Nested Lists

````python
students = [
    ["Rahul", 85],
    ["Shilpa", 92],
    ["Amit", 78]
]

for student in students:
    print(student)
````

## Access Individual Values Using a Loop

````python
students = [
    ["Rahul", 85],
    ["Shilpa", 92],
    ["Amit", 78]
]

for student in students:
    print("Name :", student[0])
    print("Marks:", student[1])
    print()
````

## Example: Product Details

````python
products = [
    ["Laptop", 55000],
    ["Mouse", 500],
    ["Keyboard", 1200]
]

for product in products:
    print(product[0], "-", product[1])
````

## Example: Monthly Sales

````python
sales = [
    ["January", 12000],
    ["February", 15000],
    ["March", 18000]
]

for month in sales:
    print(month[0], ":", month[1])
````

## Data Analytics Connection

Nested lists are useful for representing tabular data.

Example:

| Name | Age | City |
|------|----:|------|
| Rahul | 22 | Pune |
| Shilpa | 21 | Mumbai |

This structure is similar to a table in Excel or a DataFrame in Pandas.

## Summary

- A nested list is a list inside another list.
