## Methods Covered

- sort()

- index()
- copy()
- clear()

## 1. sort()

The `sort()` method arranges list elements in ascending order by default.

```python
numbers = [50, 20, 40, 10, 30]

numbers.sort()

print(numbers)
```

## Sort in Descending Order

Use the `reverse=True` argument to sort in descending order.

```python
numbers = [50, 20, 40, 10, 30]

numbers.sort(reverse=True)

print(numbers)
```

## 2. reverse()

The `reverse()` method reverses the order of the elements in a list.

```python
fruits = ["Apple", "Banana", "Mango"]

fruits.reverse()

print(fruits)
```

## sort() vs reverse()

- `sort()` arranges the elements.
- `reverse()` only changes the current order.

```python
numbers = [30, 10, 20]

numbers.reverse()

print(numbers)
```

## 3. count()

The `count()` method returns how many times a value appears in a list.

```python
numbers = [10, 20, 10, 30, 10]

print(numbers.count(10))
```

## 4. index()

The `index()` method returns the position of the first occurrence of a value.

```python
fruits = ["Apple", "Banana", "Mango", "Banana"]

print(fruits.index("Banana"))
```

## 5. copy()

The `copy()` method creates a copy of a list.

```python
original = [10, 20, 30]

copied = original.copy()

print(copied)
```

## 6. clear()

The `clear()` method removes all elements from the list.

```python
numbers = [10, 20, 30]

numbers.clear()

print(numbers)
```

## Real-Life Example

```python
marks = [75, 85, 90, 75, 60]

print("Original:", marks)

marks.sort()

print("Sorted:", marks)

print("75 appears", marks.count(75), "times")
```
