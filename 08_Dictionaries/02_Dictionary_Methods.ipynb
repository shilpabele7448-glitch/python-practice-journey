## 1. `get()`

The `get()` method is used to access a value using its key.

### Syntax

```python
dictionary.get(key)

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

print(student.get("name"))
print(student.get("age"))
````

## 2. `get()` vs `[]`

There is an important difference.

Using:

```python
student["salary"]

````python
student = {
    "name": "Rahul",
    "age": 21
}

print(student.get("salary"))
````

````python
student = {
    "name": "Rahul",
    "age": 21
}

print(student.get("salary", 0))
````

## 3. `keys()`

The `keys()` method returns all keys in the Dictionary.

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

print(student.keys())
````

## 4. `values()`

The `values()` method returns all values.

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

print(student.values())
print(list(student.values()))
````

## 5. `items()`

The `items()` method returns all key-value pairs.

Each pair is represented as a tuple:

```text
(key, value)

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

print(student.items())
````

## 6. `update()`

The `update()` method is used to:

- Add new key-value pairs
- Update existing values

````python
student = {
    "name": "Rahul",
    "age": 21
}

student.update({
    "city": "Pune",
    "age": 22
})

print(student)
````

## 7. `pop()`

The `pop()` method removes an item using its key.

### Syntax

```python
dictionary.pop(key)

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

removed = student.pop("age")

print("Removed:", removed)
print(student)
````

## 8. `popitem()`

The `popitem()` method removes and returns the **last inserted key-value pair**.

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

removed = student.popitem()

print("Removed:", removed)
print(student)
````

## 9. `setdefault()`

`setdefault()` checks whether a key exists.

If the key exists:
- It returns its existing value.
- It does not change the value.

If the key does not exist:
- It adds the key with the specified default value.

````python
student = {
    "name": "Rahul",
    "age": 21
}

result = student.setdefault("city", "Pune")

print(result)
print(student)
````

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Mumbai"
}

result = student.setdefault("city", "Pune")

print(result)
print(student)
````

## 10. `clear()`

The `clear()` method removes all items from the Dictionary.

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

student.clear()

print(student)
````

## 11. `copy()`

The `copy()` method creates a copy of a Dictionary.

````python
student = {
    "name": "Rahul",
    "age": 21
}

student_copy = student.copy()

print(student)
print(student_copy)
````

## 12. `fromkeys()`

`fromkeys()` creates a new Dictionary using a sequence of keys.

### Syntax

```python
dict.fromkeys(keys, value)

````python
keys = ["name", "age", "city"]

student = dict.fromkeys(keys)

print(student)
````

````python
keys = ["name", "age", "city"]

student = dict.fromkeys(keys, "Unknown")

print(student)
````

# 13. Data Analytics Example

Let's use Dictionary methods with product information.

````python
product = {
    "product_id": 101,
    "name": "Laptop",
    "category": "Electronics",
    "price": 55000
}

print("Keys:", product.keys())
print("Values:", product.values())
print("Items:", product.items())
````

````python
print(product.get("price"))
print(product.get("discount", 0))
````
