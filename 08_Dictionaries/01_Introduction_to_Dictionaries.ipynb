## 1. What is a Dictionary?

A Dictionary is a Python data structure that stores data in **key-value pairs**.

### Example

```python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

print(student)
````

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

print(type(student))
````

## 3. Accessing Values

We can access a Dictionary value using its key.

### Syntax

```python
dictionary[key]

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

print(student["name"])
print(student["age"])
print(student["city"])
````

### Important

Unlike Lists, Dictionaries are accessed using **keys**.

List:

```python
numbers[0]

````python
person = {
    "name": "Amit",
    "age": 25,
    "salary": 45000.50,
    "is_employee": True
}

print(person)
````

## 4. Adding Items

We can add a new key-value pair by assigning a value to a new key.

````python
student = {
    "name": "Rahul",
    "age": 21
}

student["city"] = "Pune"

print(student)
````

## 5. Updating Values

If the key already exists, assigning a new value will update the existing value.

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

student["age"] = 22

print(student)
````

## 6. Finding Dictionary Length

Use:

```python
len(dictionary)

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

print(len(student))
````

## 7. Checking Whether a Key Exists

Use the `in` operator.

```python
"name" in student

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

print("name" in student)
print("salary" in student)
````

## 8. Checking Whether a Value Exists

By default, `in` checks **keys**, not values.

To check values, we can use:

```python
value in dictionary.values()

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

print("Rahul" in student.values())
print("Mumbai" in student.values())
````

## 9. Empty Dictionary

We can create an empty Dictionary using:

```python
{}

````python
student = {}

print(student)
print(type(student))
````

````python
student = dict()

print(student)
print(type(student))
````

## 10. Removing Items

We can remove a key-value pair using the `del` keyword.

### Syntax

```python
del dictionary[key]

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

del student["city"]

print(student)
````

## 11. Removing Items Using `pop()`

The `pop()` method removes a key-value pair and returns its value.

### Syntax

```python
dictionary.pop(key)

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

removed_value = student.pop("age")

print("Removed value:", removed_value)
print(student)
````

## 12. Getting All Keys

Use:

```python
dictionary.keys()

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

print(student.keys())
````

## 13. Getting All Values

Use:

```python
dictionary.values()

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

print(student.values())
````

## 14. Getting Key-Value Pairs

Use:

```python
dictionary.items()

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

print(student.items())
````

# 15. Employee Information

Dictionaries are useful for representing structured information.

For example, an employee can have:

- Employee ID
- Name
- Department
- Salary
- Experience

````python
employee = {
    "employee_id": 101,
    "name": "Amit",
    "department": "Sales",
    "salary": 50000,
    "experience": 3
}

print(employee)
````

````python
print("Employee ID:", employee["employee_id"])
print("Name:", employee["name"])
print("Department:", employee["department"])
print("Salary:", employee["salary"])
print("Experience:", employee["experience"])
````
