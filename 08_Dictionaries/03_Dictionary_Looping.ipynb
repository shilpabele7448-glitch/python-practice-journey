## 1. Loop Through a Dictionary

When we directly use a `for` loop on a Dictionary,
the loop goes through the **keys** by default.

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

for key in student:
    print(key)
````

## 2. Using `keys()`

We can explicitly use the `keys()` method.

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

for key in student.keys():
    print(key)
````

## 3. Loop Through Values

Use the `values()` method to loop through only the values.

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

for value in student.values():
    print(value)
````

## 4. Loop Through Key-Value Pairs

Use `items()` when we need both the key and value.

We can use two variables:

```python
for key, value in dictionary.items():

````python
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Pune"
}

for key, value in student.items():
    print(key, ":", value)
````

````python
employee = {
    "name": "Amit",
    "department": "Sales",
    "salary": 50000
}

for key, value in employee.items():
    print(f"{key}: {value}")
````

## 5. Using Conditions

We can use `if` inside a Dictionary loop.

Example:

Print only employees whose salary is greater than 50000.

````python
employees = {
    "Amit": 50000,
    "Rahul": 60000,
    "Priya": 45000,
    "Neha": 70000
}

for name, salary in employees.items():
    if salary > 50000:
        print(name, salary)
````

## 6. Count Employees

Let's count how many employees have a salary greater than 50000.

````python
employees = {
    "Amit": 50000,
    "Rahul": 60000,
    "Priya": 45000,
    "Neha": 70000
}

count = 0

for salary in employees.values():
    if salary > 50000:
        count += 1

print("Employees:", count)
````

## 7. Calculate Total Salary

````python
employees = {
    "Amit": 50000,
    "Rahul": 60000,
    "Priya": 45000,
    "Neha": 70000
}

total_salary = 0

for salary in employees.values():
    total_salary += salary

print("Total salary:", total_salary)
````

## 8. Calculate Average Salary

Formula:

Average = Total Salary / Number of Employees

````python
employees = {
    "Amit": 50000,
    "Rahul": 60000,
    "Priya": 45000,
    "Neha": 70000
}

total_salary = 0

for salary in employees.values():
    total_salary += salary

average_salary = total_salary / len(employees)

print("Average salary:", average_salary)
````

## 9. Find Highest Salary

We can use a loop to find the highest salary.

Python also provides `max()`, but here we are practicing the logic.

````python
employees = {
    "Amit": 50000,
    "Rahul": 60000,
    "Priya": 45000,
    "Neha": 70000
}

highest_salary = None

for salary in employees.values():
    if highest_salary is None or salary > highest_salary:
        highest_salary = salary

print("Highest salary:", highest_salary)
````

## 10. Find Employee with Highest Salary

Now we don't just want the salary.

We want the **employee name and salary**.

````python
employees = {
    "Amit": 50000,
    "Rahul": 60000,
    "Priya": 45000,
    "Neha": 70000
}

highest_salary = None
highest_employee = None

for name, salary in employees.items():
    if highest_salary is None or salary > highest_salary:
        highest_salary = salary
        highest_employee = name

print("Employee:", highest_employee)
print("Salary:", highest_salary)
````

## 11. Find Employee with Lowest Salary

````python
employees = {
    "Amit": 50000,
    "Rahul": 60000,
    "Priya": 45000,
    "Neha": 70000
}

lowest_salary = None
lowest_employee = None

for name, salary in employees.items():
    if lowest_salary is None or salary < lowest_salary:
        lowest_salary = salary
        lowest_employee = name

print("Employee:", lowest_employee)
print("Salary:", lowest_salary)
````

## 12. Filter Dictionary Data

Create a new Dictionary containing employees whose salary is greater than 50000.

````python
employees = {
    "Amit": 50000,
    "Rahul": 60000,
    "Priya": 45000,
    "Neha": 70000
}

high_salary = {}

for name, salary in employees.items():
    if salary > 50000:
        high_salary[name] = salary

print(high_salary)
````

## 13. Increase Every Salary

Suppose the company gives every employee a 10% salary increase.

Create a new Dictionary containing the updated salaries.

````python
employees = {
    "Amit": 50000,
    "Rahul": 60000,
    "Priya": 45000,
    "Neha": 70000
}

updated_salaries = {}

for name, salary in employees.items():
    updated_salaries[name] = salary * 1.10

print(updated_salaries)
````

## 14. Create a Dictionary from Two Lists

We have:

- Employee names
- Salaries

Create a Dictionary where:

```text
Name → Salary

````python
names = ["Amit", "Rahul", "Priya", "Neha"]
salaries = [50000, 60000, 45000, 70000]

employees = {}

for i in range(len(names)):
    employees[names[i]] = salaries[i]

print(employees)
````

Suppose we have product sales:

```python
sales = {

    "Headphones": 30000
}

````python
sales = {
    "Laptop": 120000,
    "Mobile": 90000,
    "Tablet": 50000,
    "Headphones": 30000
}

for product, amount in sales.items():
    print(product, ":", amount)
````

````python
sales = {
    "Laptop": 120000,
    "Mobile": 90000,
    "Tablet": 50000,
    "Headphones": 30000
}

for product, amount in sales.items():
    if amount > 50000:
        print(product, amount)
````

````python
sales = {
    "Laptop": 120000,
    "Mobile": 90000,
    "Tablet": 50000,
    "Headphones": 30000
}

total_sales = 0

for amount in sales.values():
    total_sales += amount

print("Total sales:", total_sales)
````

## 16. Find Highest-Selling Product

Find the product that generated the highest sales.

````python
sales = {
    "Laptop": 120000,
    "Mobile": 90000,
    "Tablet": 50000,
    "Headphones": 30000
}

highest_sales = None
best_product = None

for product, amount in sales.items():
    if highest_sales is None or amount > highest_sales:
        highest_sales = amount
        best_product = product

print("Best-selling product:", best_product)
print("Sales:", highest_sales)
````

# 17. Nested Dictionaries

A Dictionary can contain another Dictionary as its value.

Example:

```python
employees = {
    "E101": {
        "name": "Amit",
        "salary": 50000
    },
    "E102": {
        "name": "Rahul",
        "salary": 60000
    }
}

````python
employees = {
    "E101": {
        "name": "Amit",
        "salary": 50000
    },
    "E102": {
        "name": "Rahul",
        "salary": 60000
    }
}

print(employees)
````

````python
for employee_id, details in employees.items():
    print("Employee ID:", employee_id)
    print("Name:", details["name"])
    print("Salary:", details["salary"])
    print()
````
