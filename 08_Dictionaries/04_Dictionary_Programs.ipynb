## Program 7 — Calculate Total

Calculate the total of all values in the Dictionary.

````python
numbers = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}

total = 0

for value in numbers.values():
    total += value

print("Total:", total)
````

## Program 8 — Calculate Average

Calculate the average of all numeric values.

````python
marks = {
    "Maths": 80,
    "Science": 75,
    "English": 90,
    "History": 85
}

total = 0

for mark in marks.values():
    total += mark

average = total / len(marks)

print("Average:", average)
````

## Program 9 — Find Maximum Value

Find the highest value in the Dictionary without using `max()`.

````python
marks = {
    "Maths": 80,
    "Science": 75,
    "English": 90,
    "History": 85
}

highest = None

for mark in marks.values():
    if highest is None or mark > highest:
        highest = mark

print("Highest mark:", highest)
````

## Program 10 — Find Subject with Highest Marks

Find both:

- Subject
- Marks

of the student with the highest marks.

````python
marks = {
    "Maths": 80,
    "Science": 75,
    "English": 90,
    "History": 85
}

highest = None
highest_subject = None

for subject, mark in marks.items():
    if highest is None or mark > highest:
        highest = mark
        highest_subject = subject

print("Subject:", highest_subject)
print("Marks:", highest)
````

## Program 11 — Count Values Greater Than 50

Count how many subjects have marks greater than 50.

````python
marks = {
    "Maths": 80,
    "Science": 45,
    "English": 90,
    "History": 35
}

count = 0

for mark in marks.values():
    if mark > 50:
        count += 1

print("Subjects above 50:", count)
````

## Program 12 — Filter Data

Create a new Dictionary containing only subjects with marks greater than 50.

````python
marks = {
    "Maths": 80,
    "Science": 45,
    "English": 90,
    "History": 35
}

passed_subjects = {}

for subject, mark in marks.items():
    if mark > 50:
        passed_subjects[subject] = mark

print(passed_subjects)
````

## Program 13 — Find Even Values

Create a new Dictionary containing only key-value pairs where the value is even.

````python
numbers = {
    "a": 10,
    "b": 15,
    "c": 20,
    "d": 25,
    "e": 30
}

even_numbers = {}

for key, value in numbers.items():
    if value % 2 == 0:
        even_numbers[key] = value

print(even_numbers)
````

## Program 14 — Increase Values by 10%

Increase every salary by 10% and store the result in a new Dictionary.

## Program 16 — Count Character Frequency

Given a string, count how many times each character appears.

Example:

```text
banana

````python
text = "banana"

frequency = {}

for character in text:

    if character in frequency:
        frequency[character] += 1
    else:
        frequency[character] = 1

print(frequency)
````

## Program 17 — Count Word Frequency

Count how many times each word appears in a sentence.

````python
sentence = "python sql python excel sql python"

words = sentence.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
````

## Program 18 — Employee Salary Analysis

Given employee salaries:

Find:

1. Total salary
2. Average salary
3. Highest salary
4. Lowest salary
5. Employees earning more than ₹50,000

````python
employees = {
    "Amit": 50000,
    "Rahul": 65000,
    "Priya": 45000,
    "Neha": 75000,
    "Rohan": 55000
}

total = 0
highest = None
lowest = None
highest_employee = None
lowest_employee = None

for name, salary in employees.items():

    total += salary

    if highest is None or salary > highest:
        highest = salary
        highest_employee = name

    if lowest is None or salary < lowest:
        lowest = salary
        lowest_employee = name

print("Total salary:", total)
print("Average salary:", total / len(employees))
print("Highest:", highest_employee, highest)
print("Lowest:", lowest_employee, lowest)

print("\nEmployees earning more than 50000:")

for name, salary in employees.items():
    if salary > 50000:
        print(name, salary)
````

## Program 19 — Employee Analysis Using Nested Dictionary

Given employee information:

Find:

1. Employees from IT
2. Employees earning more than ₹60,000
3. Highest-paid employee
4. Average salary

````python
employees = {
    "E101": {
        "name": "Amit",
        "department": "Sales",
        "salary": 50000
    },
    "E102": {
        "name": "Rahul",
        "department": "IT",
        "salary": 70000
    },
    "E103": {
        "name": "Priya",
        "department": "HR",
        "salary": 55000
    },
    "E104": {
        "name": "Neha",
        "department": "IT",
        "salary": 80000
    }
}

total_salary = 0
highest_salary = None
highest_employee = None

print("IT Employees:")

for employee_id, details in employees.items():

    total_salary += details["salary"]

    if details["department"] == "IT":
        print(details["name"])

    if details["salary"] > 60000:
        print(details["name"], "earns more than 60000")

    if highest_salary is None or details["salary"] > highest_salary:
        highest_salary = details["salary"]
        highest_employee = details["name"]

print("\nHighest-paid employee:", highest_employee)
print("Highest salary:", highest_salary)
print("Average salary:", total_salary / len(employees))
````

# Program 20 — Student Result Analysis

A student has marks in five subjects.

Calculate:

1. Total marks
2. Average marks
3. Highest marks
4. Lowest marks
5. Highest-scoring subject
6. Number of subjects passed
7. Overall result

Passing mark = 40


````python
marks = {
    "Maths": 78,
    "Science": 65,
    "English": 82,
    "History": 55,
    "Geography": 35
}

total = 0
highest = None
lowest = None
highest_subject = None
lowest_subject = None
passed_subjects = 0

for subject, mark in marks.items():

    total += mark

    if highest is None or mark > highest:
        highest = mark
        highest_subject = subject

    if lowest is None or mark < lowest:
        lowest = mark
        lowest_subject = subject

    if mark >= 40:
        passed_subjects += 1

average = total / len(marks)

print("===== STUDENT RESULT =====")
print("Total:", total)
print("Average:", average)
print("Highest:", highest_subject, highest)
print("Lowest:", lowest_subject, lowest)
print("Passed subjects:", passed_subjects)

if passed_subjects == len(marks):
    print("Overall Result: PASS")
else:
    print("Overall Result: FAIL")
````
