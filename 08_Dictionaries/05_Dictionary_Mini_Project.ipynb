## 1. Employee Dataset

We will represent each employee using a nested Dictionary.

The Employee ID will be the main key.

````python
employees = {
    "E101": {
        "name": "Amit",
        "department": "Sales",
        "salary": 50000,
        "experience": 2
    },
    
    "E102": {
        "name": "Rahul",
        "department": "IT",
        "salary": 70000,
        "experience": 4
    },
    
    "E103": {
        "name": "Priya",
        "department": "HR",
        "salary": 55000,
        "experience": 3
    },
    
    "E104": {
        "name": "Neha",
        "department": "IT",
        "salary": 80000,
        "experience": 5
    },
    
    "E105": {
        "name": "Rohan",
        "department": "Finance",
        "salary": 65000,
        "experience": 4
    },
    
    "E106": {
        "name": "Sneha",
        "department": "Sales",
        "salary": 48000,
        "experience": 2
    }
}

print(employees)
````

## 2. Display All Employees

Loop through the Dictionary and display each employee's information.

````python
for employee_id, details in employees.items():
    print("Employee ID:", employee_id)
    print("Name:", details["name"])
    print("Department:", details["department"])
    print("Salary:", details["salary"])
    print("Experience:", details["experience"])
    print("-" * 30)
````

## 3. Total Number of Employees

````python
total_employees = len(employees)

print("Total employees:", total_employees)
````

## 4. Total Salary Expense

Calculate the total salary the company pays to all employees.

````python
total_salary = 0

for details in employees.values():
    total_salary += details["salary"]

print("Total salary expense:", total_salary)
````

## 5. Average Salary

````python
average_salary = total_salary / total_employees

print("Average salary:", average_salary)
````

## 6. Highest-Paid Employee

Find:

- Employee name
- Salary

````python
highest_salary = None
highest_employee = None

for details in employees.values():
    
    if highest_salary is None or details["salary"] > highest_salary:
        highest_salary = details["salary"]
        highest_employee = details["name"]

print("Highest-paid employee:", highest_employee)
print("Salary:", highest_salary)
````

## 7. Lowest-Paid Employee

````python
lowest_salary = None
lowest_employee = None

for details in employees.values():
    
    if lowest_salary is None or details["salary"] < lowest_salary:
        lowest_salary = details["salary"]
        lowest_employee = details["name"]

print("Lowest-paid employee:", lowest_employee)
print("Salary:", lowest_salary)
````

## 8. High-Salary Employees

Find employees whose salary is greater than ₹60,000.

````python
print("Employees earning above ₹60,000:")

for details in employees.values():
    
    if details["salary"] > 60000:
        print(details["name"], "→", details["salary"])
````

## 9. Experienced Employees

Find employees who have more than 3 years of experience.

````python
print("Employees with more than 3 years experience:")

for details in employees.values():
    
    if details["experience"] > 3:
        print(details["name"], "→", details["experience"], "years")
````

## 10. IT Department Employees

Find all employees who belong to the IT department.

````python
print("IT Employees:")

for details in employees.values():
    
    if details["department"] == "IT":
        print(details["name"])
````

## 11. Department-Wise Employee Count

Find how many employees work in each department.

Expected structure:

```text
Sales → 2
IT → 2
HR → 1
Finance → 1

````python
department_count = {}

for details in employees.values():
    
    department = details["department"]
    
    if department in department_count:
        department_count[department] += 1
    else:
        department_count[department] = 1

print(department_count)
````

## 12. Department-Wise Salary

Calculate the total salary paid by each department.

Expected idea:

```text
Sales → total Sales salaries
IT → total IT salaries
HR → total HR salaries
Finance → total Finance salaries

````python
department_salary = {}

for details in employees.values():
    
    department = details["department"]
    salary = details["salary"]
    
    if department in department_salary:
        department_salary[department] += salary
    else:
        department_salary[department] = salary

print(department_salary)
````

## 13. Department-Wise Average Salary

Now calculate the average salary for every department.

Formula:

```text
Average Salary =
Total Department Salary / Number of Employees

````python
department_count = {}
department_salary = {}

for details in employees.values():
    
    department = details["department"]
    salary = details["salary"]
    
    if department in department_count:
        department_count[department] += 1
        department_salary[department] += salary
    else:
        department_count[department] = 1
        department_salary[department] = salary

for department in department_salary:
    
    average = department_salary[department] / department_count[department]
    
    print(department, "→", average)
````

## 14. Department with Highest Salary Expense

Find which department has the highest total salary expense.

````python
highest_department_salary = None
highest_department = None

for department, salary in department_salary.items():
    
    if (highest_department_salary is None 
            or salary > highest_department_salary):
        
        highest_department_salary = salary
        highest_department = department

print("Department:", highest_department)
print("Salary expense:", highest_department_salary)
````

## 15. Create a New Dictionary

Create a new Dictionary containing only employees whose salary is greater than ₹60,000.

The new Dictionary should retain the employee ID and employee details.

````python
high_salary_employees = {}

for employee_id, details in employees.items():
    
    if details["salary"] > 60000:
        high_salary_employees[employee_id] = details

print(high_salary_employees)
````

## 16. Search Employee

Ask the user for an Employee ID and display the employee information.

If the Employee ID doesn't exist, display an appropriate message.

````python
employee_id = input("Enter Employee ID: ")

if employee_id in employees:
    print(employees[employee_id])
else:
    print("Employee not found")
````

## 17. Salary Increment

The company decides to increase every employee's salary by 10%.

Create a new Dictionary containing the updated employee information.

````python
updated_employees = {}

for employee_id, details in employees.items():
    
    updated_details = details.copy()
    
    updated_details["salary"] = details["salary"] * 1.10
    
    updated_employees[employee_id] = updated_details

print(updated_employees)
````

# 📊 Employee Analysis Dashboard

Let's display the most important results together.

````python
print("=" * 40)
print("       EMPLOYEE ANALYSIS")
print("=" * 40)

print("Total Employees:", total_employees)
print("Total Salary Expense:", total_salary)
print("Average Salary:", average_salary)

print("\nHighest-Paid Employee:")
print(highest_employee, "→", highest_salary)

print("\nLowest-Paid Employee:")
print(lowest_employee, "→", lowest_salary)

print("\nDepartment Employee Count:")
for department, count in department_count.items():
    print(department, "→", count)

print("\nDepartment Salary:")
for department, salary in department_salary.items():
    print(department, "→", salary)

print("=" * 40)
````
