## Program 1 — Create a Set

Create a Set containing five numbers and display it.

````python
numbers = {10, 20, 30, 40, 50}

print(numbers)
````

## Program 2 — Remove Duplicates

Convert a list containing duplicate values into a Set.

````python
numbers = [10, 20, 10, 30, 20, 40, 10]

unique_numbers = set(numbers)

print("Original:", numbers)
print("Unique:", unique_numbers)
````

## Program 3 — Count Unique Values

Find how many unique values exist in a list.

````python
numbers = [10, 20, 10, 30, 20, 40, 10]

unique_numbers = set(numbers)

print("Number of unique values:", len(unique_numbers))
````

## Program 4 — Check Whether an Element Exists

Check whether `30` exists in the Set.

````python
numbers = {10, 20, 30, 40, 50}

if 30 in numbers:
    print("30 exists in the Set")
else:
    print("30 does not exist in the Set")
````

## Program 5 — Add an Element

Add a new number to an existing Set.

````python
numbers = {10, 20, 30}

numbers.add(40)

print(numbers)
````

## Program 6 — Add Multiple Elements

Add multiple numbers using `update()`.

````python
numbers = {10, 20, 30}

numbers.update([40, 50, 60])

print(numbers)
````

## Program 7 — Remove an Element

Remove `30` from the Set.

````python
numbers = {10, 20, 30, 40}

numbers.remove(30)

print(numbers)
````

## Program 8 — Find Even Numbers

Create a new Set containing only even numbers.

````python
numbers = {10, 15, 20, 25, 30, 35, 40}

even_numbers = set()

for number in numbers:
    if number % 2 == 0:
        even_numbers.add(number)

print(even_numbers)
````

## Program 9 — Find Odd Numbers

Create a new Set containing only odd numbers.

````python
numbers = {10, 15, 20, 25, 30, 35, 40}

odd_numbers = set()

for number in numbers:
    if number % 2 != 0:
        odd_numbers.add(number)

print(odd_numbers)
````

## Program 10 — Find Union

Find all unique elements from two Sets.

````python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

result = set1.union(set2)

print("Union:", result)
````

## Program 11 — Find Common Elements

Find elements that are present in both Sets.

````python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

common = set1.intersection(set2)

print("Common elements:", common)
````

## Program 12 — Find Difference

Find elements present in `set1` but not in `set2`.

````python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

result = set1.difference(set2)

print("Set1 only:", result)
````

## Program 13 — Symmetric Difference

Find elements that exist in either Set but not in both.

````python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

result = set1.symmetric_difference(set2)

print("Symmetric difference:", result)
````

## Program 14 — Check Subset

Check whether one Set is completely contained inside another Set.

````python
set1 = {1, 2}
set2 = {1, 2, 3, 4, 5}

if set1.issubset(set2):
    print("set1 is a subset of set2")
else:
    print("set1 is not a subset of set2")
````

## Program 15 — Check Superset

````python
set1 = {1, 2, 3, 4, 5}
set2 = {1, 2}

if set1.issuperset(set2):
    print("set1 is a superset of set2")
else:
    print("set1 is not a superset of set2")
````

## Program 16 — Check Disjoint Sets

Check whether two Sets have no common elements.

````python
set1 = {1, 2, 3}
set2 = {4, 5, 6}

if set1.isdisjoint(set2):
    print("The Sets are disjoint")
else:
    print("The Sets are not disjoint")
````

## Program 17 — Find Common Customers

A company has two groups of customers:

- Online customers
- Store customers

Find customers who purchased through both channels.

````python
online_customers = {101, 102, 103, 104, 105}
store_customers = {103, 104, 106, 107}

common_customers = online_customers.intersection(store_customers)

print("Customers using both channels:")
print(common_customers)
````

## Program 18 — Find Online-Only Customers

Find customers who purchased online but did not purchase from the physical store.

````python
online_customers = {101, 102, 103, 104, 105}
store_customers = {103, 104, 106, 107}

online_only = online_customers - store_customers

print("Online-only customers:")
print(online_only)
````

## Program 19 — Find Unique Skills

Two groups of employees have different technical skills.

Find all unique skills across both groups.

````python
group1 = {"Python", "SQL", "Excel"}
group2 = {"Python", "Power BI", "Statistics"}

all_skills = group1.union(group2)

print("All unique skills:")
print(all_skills)
````

# Program 20 — Complete Customer Analysis

A company has two customer groups:

```python
online_customers = {101, 102, 103, 104, 105}
store_customers = {103, 104, 106, 107}

````python
online_customers = {101, 102, 103, 104, 105}
store_customers = {103, 104, 106, 107}

# 1. Total unique customers
all_customers = online_customers | store_customers

# 2. Customers using both channels
both_channels = online_customers & store_customers

# 3. Online-only customers
online_only = online_customers - store_customers

# 4. Store-only customers
store_only = store_customers - online_customers

# 5. Customers using exactly one channel
exactly_one = online_customers ^ store_customers

print("Total unique customers:", len(all_customers))
print("All customers:", all_customers)

print("Both channels:", both_channels)
print("Online only:", online_only)
print("Store only:", store_only)
print("Exactly one channel:", exactly_one)
````
