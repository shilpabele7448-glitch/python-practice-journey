## 1. Creating a Set

Example:

```python
numbers = {10, 20, 30}

````python
numbers = {10, 20, 30}

print(numbers)
print(type(numbers))
````

## 2. add()

The `add()` method adds **one element** to a Set.

### Syntax

```python
set_name.add(element)

````python
numbers = {10, 20, 30}

numbers.add(40)

print(numbers)
````

````python
fruits = {"Apple", "Banana"}

fruits.add("Mango")

print(fruits)
````

### What happens if we add a duplicate?

Sets do not allow duplicates.

So adding an existing element does not create another copy.

````python
numbers = {10, 20, 30}

numbers.add(20)

print(numbers)
````

## 3. update()

The `update()` method is used to add **multiple elements** to a Set.

### Syntax

```python
set_name.update(iterable)

````python
numbers = {10, 20, 30}

numbers.update([40, 50, 60])

print(numbers)
````

````python
fruits = {"Apple", "Banana"}

fruits.update(["Mango", "Orange", "Grapes"])

print(fruits)
````

### `add()` vs `update()`

| Method | Purpose |
|---|---|
| `add()` | Adds one element |
| `update()` | Adds multiple elements |

Example:

```python
numbers.add(40)

## 4. remove()

The `remove()` method removes a specific element from a Set.

### Syntax

```python
set_name.remove(element)

````python
numbers = {10, 20, 30, 40}

numbers.remove(30)

print(numbers)
````

### Important

If the element does not exist, `remove()` raises a `KeyError`.

Example:

```python
numbers.remove(100)

````python
numbers = {10, 20, 30}

numbers.remove(100)
````

## 5. discard()

`discard()` also removes an element from a Set.

The important difference is:

- `remove()` → error if element doesn't exist
- `discard()` → no error if element doesn't exist

````python
numbers = {10, 20, 30}

numbers.discard(20)

print(numbers)
````

````python
numbers = {10, 20, 30}

numbers.discard(100)

print(numbers)
````

## `remove()` vs `discard()`

| Method | Element exists | Element doesn't exist |
|---|---|---|
| `remove()` | Removes it | ❌ KeyError |
| `discard()` | Removes it | ✅ No error |

### Practical Tip

If you're not sure whether an element exists, `discard()` is often safer.

## 6. pop()

The `pop()` method removes and returns an element from a Set.

Because Sets are unordered, you should **not assume which element will be removed**.

````python
numbers = {10, 20, 30, 40}

removed = numbers.pop()

print("Removed:", removed)
print("Remaining:", numbers)
````

### Important

With a List:

```python
numbers.pop()

## 7. clear()

The `clear()` method removes **all elements** from a Set.

````python
numbers = {10, 20, 30, 40}

numbers.clear()

print(numbers)
````

## 8. copy()

The `copy()` method creates a copy of a Set.

````python
numbers = {10, 20, 30}

new_numbers = numbers.copy()

print("Original:", numbers)
print("Copy:", new_numbers)
````

````python
numbers = {10, 20, 30}

new_numbers = numbers.copy()

new_numbers.add(40)

print("Original:", numbers)
print("Copy:", new_numbers)
````

## 9. Adding Multiple Elements

Use `update()` when you want to add multiple values.

You can update a Set using:

- List
- Tuple
- Another Set
- String

Example:

````python
numbers = {10, 20}

numbers.update((30, 40, 50))

print(numbers)
````

````python
numbers = {10, 20}

numbers.update({30, 40, 50})

print(numbers)
````

## `update()` with a String

Be careful when using a string with `update()`.

A string is iterable, so its characters are added individually.

````python
letters = {"A", "B"}

letters.update("CAT")

print(letters)
````

## 10. Safe Removal Using `in`

We can check whether an element exists before using `remove()`.

This avoids a `KeyError`.

````python
numbers = {10, 20, 30}

if 20 in numbers:
    numbers.remove(20)

print(numbers)
````

### Another approach

We can simply use `discard()`:

```python
numbers.discard(20)

## 11. Student Subjects

Suppose a student is enrolled in several subjects.

We can use a Set because duplicate subjects are unnecessary.

````python
subjects = {"Python", "SQL", "Excel"}

subjects.add("Power BI")

print(subjects)
````

````python
subjects = {"Python", "SQL", "Excel"}

subjects.update(["Statistics", "Machine Learning"])

print(subjects)
````

## Remove a Subject

Suppose the student stops studying Excel.

````python
subjects = {"Python", "SQL", "Excel", "Power BI"}

subjects.discard("Excel")

print(subjects)
````

## 12. Unique Customer IDs

Suppose a company has repeated customer IDs because customers made multiple purchases.

We can use a Set to find unique customers.

````python
customer_ids = [101, 102, 101, 103, 102, 104, 101]

unique_customers = set(customer_ids)

print("Unique customers:", unique_customers)
print("Number of unique customers:", len(unique_customers))
````
