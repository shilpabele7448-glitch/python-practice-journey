# 📊 Set Mini Project — Customer Purchase Analysis

## Business Problem

A company sells products through two channels:

- Online
- Physical Store

The company wants to understand customer behavior.

We will use Python Sets to find:

1. Total unique customers
2. Customers using both channels
3. Online-only customers
4. Store-only customers
5. Customers using exactly one channel
6. Number of customers in each category

### Concepts Used

- Sets
- `len()`
- Union
- Intersection
- Difference
- Symmetric Difference

**Author:** Shilpa

## Dataset

We have two groups of customer IDs.

### Online Customers

Customers who purchased through the company's website.

### Store Customers

Customers who purchased from the physical store.

````python
online_customers = {
    101, 102, 103, 104, 105,
    106, 107, 108
}

store_customers = {
    103, 104, 105, 109,
    110, 111
}

print("Online Customers:", online_customers)
print("Store Customers:", store_customers)
````

## 1. Total Unique Customers

Some customers purchased through both channels.

We don't want to count them twice.

Therefore, we use **Union**.

Union gives us all unique customers.

````python
all_customers = online_customers | store_customers

print("All unique customers:", all_customers)
print("Total unique customers:", len(all_customers))
````

## 2. Customers Using Both Channels

We want customers who purchased:

- Online
- AND
- In the physical store

This is an **Intersection** problem.

````python
both_channels = online_customers & store_customers

print("Customers using both channels:", both_channels)
print("Number of customers:", len(both_channels))
````

## 3. Online-Only Customers

We want customers who purchased online but did NOT purchase from the physical store.

Use:

```python
online_customers - store_customers

````python
online_only = online_customers - store_customers

print("Online-only customers:", online_only)
print("Number of online-only customers:", len(online_only))
````

## 4. Store-Only Customers

We want customers who purchased from the store but did NOT purchase online.

Use:

```python
store_customers - online_customers

````python
store_only = store_customers - online_customers

print("Store-only customers:", store_only)
print("Number of store-only customers:", len(store_only))
````

## 5. Customers Using Exactly One Channel

We want customers who used:

- Online only
- OR Store only

But NOT both.

This is called **Symmetric Difference**.

We can use:

```python
online_customers ^ store_customers

````python
exactly_one_channel = online_customers ^ store_customers

print("Customers using exactly one channel:")
print(exactly_one_channel)

print("Number of customers:", len(exactly_one_channel))
````

## 6. Customer Analysis Summary

Let's calculate all important metrics together.

````python
total_customers = len(all_customers)
both = len(both_channels)
online_only_count = len(online_only)
store_only_count = len(store_only)
exactly_one_count = len(exactly_one_channel)

print("===== CUSTOMER ANALYSIS =====")
print("Total unique customers:", total_customers)
print("Customers using both channels:", both)
print("Online-only customers:", online_only_count)
print("Store-only customers:", store_only_count)
print("Exactly one channel:", exactly_one_count)
````

## 7. Verify the Results

Every unique customer should belong to one of these categories:

- Both channels
- Online only
- Store only

Let's verify our calculation.

````python
calculated_total = (
    len(both_channels)
    + len(online_only)
    + len(store_only)
)

print("Total unique customers:", total_customers)
print("Calculated total:", calculated_total)

print("Result:", total_customers == calculated_total)
````

### Question 6
Which customer group is larger:
Online-only or Store-only?

````python
print("Total unique customers:", len(all_customers))

print("Both channels:", len(both_channels))

print("Online only:", len(online_only))

print("Store only:", len(store_only))

print("Exactly one channel:", len(exactly_one_channel))

if len(online_only) > len(store_only):
    print("Online-only customers are larger")
elif len(store_only) > len(online_only):
    print("Store-only customers are larger")
else:
    print("Both groups have the same number of customers")
````

## 10. Classify Customers

Now let's classify every customer into one of three groups:

- Both Channels
- Online Only
- Store Only

This is closer to the type of thinking used in Data Analytics.

````python
for customer in all_customers:

    if customer in online_customers and customer in store_customers:
        print(customer, "→ Both Channels")

    elif customer in online_customers:
        print(customer, "→ Online Only")

    else:
        print(customer, "→ Store Only")
````
