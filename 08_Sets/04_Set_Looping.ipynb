## 1. Basic `for` Loop

We can use a `for` loop to access each element of a Set.

Example:

```python
numbers = {10, 20, 30, 40}

for number in numbers:
    print(number)

````python
numbers = {10, 20, 30, 40}

for number in numbers:
    print(number)
````

## 2. Loop Through a Set of Strings

````python
fruits = {"Apple", "Banana", "Mango", "Orange"}

for fruit in fruits:
    print(fruit)
````

## 3. Using `if` with a Set

We can use conditions inside a loop to filter elements.

Example:

Find numbers greater than 20.

````python
numbers = {10, 20, 30, 40, 50}

for number in numbers:
    if number > 20:
        print(number)
````

## 4. Find Even Numbers

A number is even when:

```python
number % 2 == 0

````python
numbers = {10, 15, 20, 25, 30, 35, 40}

for number in numbers:
    if number % 2 == 0:
        print(number)
````

## 5. Find Odd Numbers

````python
numbers = {10, 15, 20, 25, 30, 35, 40}

for number in numbers:
    if number % 2 != 0:
        print(number)
````

## 6. Count Elements Using a Loop

We can use a counter variable.

Start with:

```python
count = 0

````python
numbers = {10, 20, 30, 40, 50}

count = 0

for number in numbers:
    count += 1

print("Number of elements:", count)
````

## 7. Calculate the Sum

We can calculate the total manually using a loop.

````python
numbers = {10, 20, 30, 40}

total = 0

for number in numbers:
    total += number

print("Total:", total)
````

## 8. Calculate Average

Formula:

```text
Average = Total / Number of Elements

````python
numbers = {10, 20, 30, 40}

total = 0

for number in numbers:
    total += number

average = total / len(numbers)

print("Average:", average)
````

## 9. Find the Maximum Value

We can start with the first element obtained from the Set using `next(iter())`.

Then compare every value with the current maximum.

````python
numbers = {10, 20, 30, 40, 50}

maximum = next(iter(numbers))

for number in numbers:
    if number > maximum:
        maximum = number

print("Maximum:", maximum)
````

## 10. Find the Minimum Value

````python
numbers = {10, 20, 30, 40, 50}

minimum = next(iter(numbers))

for number in numbers:
    if number < minimum:
        minimum = number

print("Minimum:", minimum)
````

## 11. Search for an Element

We can use a loop and a condition to search for a particular value.

````python
numbers = {10, 20, 30, 40, 50}

search_value = 30
found = False

for number in numbers:
    if number == search_value:
        found = True
        break

if found:
    print("Value found")
else:
    print("Value not found")
````

## 12. Create a New Set Using a Loop

Suppose we want to create a Set containing only even numbers.

````python
numbers = {10, 15, 20, 25, 30, 35, 40}

even_numbers = set()

for number in numbers:
    if number % 2 == 0:
        even_numbers.add(number)

print(even_numbers)
````

## 13. Create a Set of Squares

We can create a new Set containing the square of each number.

````python
numbers = {1, 2, 3, 4, 5}

squares = set()

for number in numbers:
    squares.add(number ** 2)

print(squares)
````

## 14. Filter Values

Find all numbers greater than 25.

````python
numbers = {10, 15, 20, 25, 30, 35, 40}

greater_than_25 = set()

for number in numbers:
    if number > 25:
        greater_than_25.add(number)

print(greater_than_25)
````

## 15. Find Long Words

Find words containing more than 5 characters.

````python
words = {"Python", "SQL", "Analytics", "Excel", "PowerBI"}

long_words = set()

for word in words:
    if len(word) > 5:
        long_words.add(word)

print(long_words)
````

## 16. Nested Loops with Sets

We can use nested loops when working with multiple Sets.

Example:

Compare every element from Set A with every element from Set B.

````python
A = {1, 2, 3}
B = {3, 4, 5}

for a in A:
    for b in B:
        print(a, b)
````

## 17. Customer Analysis

Suppose we have customer IDs from two sales channels.

We want to find customers who purchased online.

````python
online_customers = {101, 102, 103, 104, 105}

print("Online Customers:")

for customer in online_customers:
    print(customer)
````

## Find Customers with Specific IDs

Suppose we want to find customers whose ID is greater than 102.

````python
online_customers = {101, 102, 103, 104, 105}

for customer in online_customers:
    if customer > 102:
        print(customer)
````

## 18. Product Category Analysis

Find categories whose names contain more than 6 characters.

````python
categories = {
    "Electronics",
    "Clothing",
    "Grocery",
    "Furniture",
    "Books"
}

for category in categories:
    if len(category) > 6:
        print(category)
````
