## What is a Set?

A set is a collection of unique elements.

Sets are written using curly brackets:

```python
{10, 20, 30}

````python
numbers = {10, 20, 30, 40}

print(numbers)
print(type(numbers))
````

## Creating a Set

We can create a set by placing values inside curly brackets `{}`.

````python
fruits = {"Apple", "Banana", "Mango", "Orange"}

print(fruits)
````

## Sets Automatically Remove Duplicates

Sets store only unique values.

If we provide duplicate values, Python keeps only one copy.

````python
numbers = {10, 20, 10, 30, 20, 40, 10}

print(numbers)
````

## Why Are Sets Useful?

Suppose we have:

```python
numbers = [10, 20, 10, 30, 20, 40]

````python
numbers = [10, 20, 10, 30, 20, 40]

unique_numbers = set(numbers)

print(unique_numbers)
````

## Sets Are Unordered

A set does not maintain elements using indexes in the same way as a list or tuple.

For example:

```python
numbers = {10, 20, 30, 40}

````python
numbers = {10, 20, 30, 40}

print(numbers)
````

````python
numbers = {10, 20, 30, 40}

print(numbers[0])
````

### Important

Lists and tuples support indexing:

```python
my_list[0]
my_tuple[0]

## Sets Are Mutable

A set itself can be changed after it is created.

We will learn how to add and remove elements in the next lessons.

````python
numbers = {10, 20, 30}

numbers.add(40)

print(numbers)
````

## Sets Can Store Different Data Types

A set can contain different types of immutable values.

For example:

- integers
- strings
- floats
- tuples

````python
data = {10, "Python", 3.14, (1, 2)}

print(data)
````

## Finding the Number of Elements

Use the `len()` function to find the number of unique elements in a set.

````python
numbers = {10, 20, 30, 40, 50}

print(len(numbers))
````

## Checking Whether an Element Exists

We can use:

```python
in

````python
fruits = {"Apple", "Banana", "Mango"}

print("Apple" in fruits)
print("Orange" in fruits)
````

````python
fruits = {"Apple", "Banana", "Mango"}

if "Mango" in fruits:
    print("Mango is available")
else:
    print("Mango is not available")
````

## List vs Tuple vs Set

| Feature | List | Tuple | Set |
|---|---|---|---|
| Ordered | Yes | Yes | No |
| Allows duplicates | Yes | Yes | No |
| Mutable | Yes | No | Yes |
| Indexing | Yes | Yes | No |
| Syntax | `[]` | `()` | `{}` |
| Main use | Collection of changing data | Fixed data | Unique values |

Example:

```python
my_list = [10, 20, 10]
my_tuple = (10, 20, 10)
my_set = {10, 20, 10}

## Sets in Data Analytics

Sets are very useful when we need unique values.

For example, suppose we have customer cities:

```python
cities = (
    "Mumbai",
    "Pune",
    "Mumbai",
    "Nashik",
    "Pune",
    "Nagpur"
)

````python
cities = (
    "Mumbai",
    "Pune",
    "Mumbai",
    "Nashik",
    "Pune",
    "Nagpur"
)

unique_cities = set(cities)

print(unique_cities)
print("Number of unique cities:", len(unique_cities))
````
