## 1. Union

Union combines all unique elements from two Sets.

Suppose:

```python
A = {1, 2, 3}
B = {3, 4, 5}

````python
A = {1, 2, 3}
B = {3, 4, 5}

result = A.union(B)

print(result)
````

### Union Using `|`

We can also use the `|` operator.

````python
A = {1, 2, 3}
B = {3, 4, 5}

result = A | B

print(result)
````

### `union()` vs `|`

Both produce the same result:

```python
A.union(B)

## 2. Intersection

Intersection finds the elements that are present in **both Sets**.

Example:

```python
A = {1, 2, 3}
B = {3, 4, 5}

````python
A = {1, 2, 3}
B = {3, 4, 5}

result = A.intersection(B)

print(result)
````

### Intersection Using `&`

We can also use:

```python
A & B

````python
A = {1, 2, 3}
B = {3, 4, 5}

result = A & B

print(result)
````

## 3. Difference

Difference finds elements that are present in the **first Set but not in the second Set**.

Example:

```python
A = {1, 2, 3}
B = {3, 4, 5}

````python
A = {1, 2, 3}
B = {3, 4, 5}

result = A.difference(B)

print(result)
````

````python
A = {1, 2, 3}
B = {3, 4, 5}

print(A - B)
````

### Important: Difference Is Directional

These are not necessarily the same:

```python
A - B

````python
A = {1, 2, 3}
B = {3, 4, 5}

print("A - B:", A - B)
print("B - A:", B - A)
````

## 4. Symmetric Difference

Symmetric difference finds elements that are present in **either Set, but not in both**.

Example:

```python
A = {1, 2, 3}
B = {3, 4, 5}

````python
A = {1, 2, 3}
B = {3, 4, 5}

result = A.symmetric_difference(B)

print(result)
````

### Symmetric Difference Using `^`

````python
A = {1, 2, 3}
B = {3, 4, 5}

print(A ^ B)
````

## 5. Subset

A Set is a **subset** of another Set if every element of the first Set exists in the second Set.

Example:

```python
A = {1, 2}
B = {1, 2, 3, 4}

````python
A = {1, 2}
B = {1, 2, 3, 4}

print(A.issubset(B))
````

````python
A = {1, 5}
B = {1, 2, 3, 4}

print(A.issubset(B))
````

### Subset Using `<=`

We can also write:

```python
A <= B

````python
A = {1, 2}
B = {1, 2, 3, 4}

print(A <= B)
````

## 6. Superset

A Set is a **superset** of another Set if it contains all elements of that Set.

Example:

```python
A = {1, 2, 3, 4}
B = {1, 2}

````python
A = {1, 2, 3, 4}
B = {1, 2}

print(A.issuperset(B))
````

### Superset Using `>=`

We can also use:

```python
A >= B

````python
A = {1, 2, 3, 4}
B = {1, 2}

print(A >= B)
````

## 7. Disjoint Sets

Two Sets are disjoint if they have **no elements in common**.

Example:

```python
A = {1, 2, 3}
B = {4, 5, 6}

````python
A = {1, 2, 3}
B = {4, 5, 6}

print(A.isdisjoint(B))
````

````python
A = {1, 2, 3}
B = {3, 4, 5}

print(A.isdisjoint(B))
````

# 8. Data Analytics Example — Customers

Suppose:

- `online_customers` = customers who purchased online
- `store_customers` = customers who purchased from a physical store

We can use Set operations to analyze customers.

````python
online_customers = {101, 102, 103, 104, 105}
store_customers = {103, 104, 106, 107}

print("Online:", online_customers)
print("Store:", store_customers)
````

### Customers Who Purchased Through Either Channel

````python
all_customers = online_customers.union(store_customers)

print(all_customers)
````

### Customers Who Used Both Channels

````python
both_channels = online_customers.intersection(store_customers)

print(both_channels)
````

### Online-Only Customers

````python
online_only = online_customers - store_customers

print(online_only)
````

### Store-Only Customers

````python
store_only = store_customers - online_customers

print(store_only)
````

# 9. Skills Analysis

Suppose two groups of employees have different technical skills.

We want to find:

- Skills common to both groups
- Skills unique to Group A
- Skills unique to Group B
- All skills

````python
group_a = {"Python", "SQL", "Excel", "Power BI"}
group_b = {"Python", "SQL", "Tableau", "Statistics"}

print("Group A:", group_a)
print("Group B:", group_b)
````

````python
print("All Skills:", group_a | group_b)
print("Common Skills:", group_a & group_b)
print("Only Group A:", group_a - group_b)
print("Only Group B:", group_b - group_a)
````
