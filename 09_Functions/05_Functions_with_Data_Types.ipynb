## Functions with Numbers

Functions can accept numbers as arguments and perform
mathematical calculations.

```python
def add(a, b):
    return a + b

print(add(10, 20))
```

```python
def calculate(a, b):
    addition = a + b
    subtraction = a - b
    multiplication = a * b

    return addition, subtraction, multiplication

result = calculate(20, 5)

print(result)
```

## Functions with Strings

Functions can accept strings and perform operations such as:

- Changing case
- Finding length
- Checking characters
- Joining strings
- Reversing strings

```python
def uppercase_text(text):
    return text.upper()

print(uppercase_text("python"))
```

```python
def string_info(text):
    return len(text), text.upper(), text.lower()

length, upper, lower = string_info("Python")

print("Length:", length)
print("Uppercase:", upper)
print("Lowercase:", lower)
```

```python
def count_characters(text):
    return len(text)

print(count_characters("Data Analytics"))
```

## Functions with Lists

Lists are very important in Data Analytics.

A function can accept a list and calculate:

- Total
- Average
- Maximum
- Minimum
- Number of elements

```python
def list_summary(numbers):
    total = sum(numbers)
    average = total / len(numbers)
    maximum = max(numbers)
    minimum = min(numbers)

    return total, average, maximum, minimum

numbers = [10, 20, 30, 40, 50]

total, average, maximum, minimum = list_summary(numbers)

print("Total:", total)
print("Average:", average)
print("Maximum:", maximum)
print("Minimum:", minimum)
```

```python
def get_even_numbers(numbers):
    even_numbers = []

    for number in numbers:
        if number % 2 == 0:
            even_numbers.append(number)

    return even_numbers

numbers = [1, 2, 3, 4, 5, 6, 7, 8]

print(get_even_numbers(numbers))
```

## Functions with Tuples

A tuple can also be passed to a function.

Tuples are useful when we want an ordered collection
that should not normally be changed.

```python
def tuple_sum(numbers):
    return sum(numbers)

numbers = (10, 20, 30, 40)

print(tuple_sum(numbers))
```

```python
def employee_details():
    return "Shilpa", "Data Analyst", 50000

name, role, salary = employee_details()

print("Name:", name)
print("Role:", role)
print("Salary:", salary)
```

## Functions with Sets

Sets contain unique values.

Functions can be used to perform:

- Union
- Intersection
- Difference
- Remove duplicates

```python
def common_values(set1, set2):
    return set1.intersection(set2)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(common_values(a, b))
```

```python
def remove_duplicates(values):
    return set(values)

numbers = [10, 20, 10, 30, 20, 40]

print(remove_duplicates(numbers))
```

## Functions with Dictionaries

Dictionaries store data in key-value pairs.

Functions can accept dictionaries and extract or calculate
information from them.

```python
def display_student(student):
    print("Name:", student["name"])
    print("Age:", student["age"])
    print("City:", student["city"])

student = {
    "name": "Rahul",
    "age": 22,
    "city": "Pune"
}

display_student(student)
```

```python
def get_salary(employee):
    return employee["salary"]

employee = {
    "name": "Amit",
    "salary": 50000
}

print(get_salary(employee))
```

```python
def calculate_total_sales(sales):
    return sum(sales.values())

sales = {
    "Monday": 10000,
    "Tuesday": 15000,
    "Wednesday": 12000,
    "Thursday": 18000
}

print("Total Sales:", calculate_total_sales(sales))
```

```python
employees = [
    {"name": "Amit", "salary": 50000},
    {"name": "Rahul", "salary": 60000},
    {"name": "Priya", "salary": 55000}
]

def total_salary(employees):
    total = 0

    for employee in employees:
        total += employee["salary"]

    return total

print("Total Salary:", total_salary(employees))
```

```python
def highest_salary(employees):
    highest = 0

    for employee in employees:
        if employee["salary"] > highest:
            highest = employee["salary"]

    return highest

print("Highest Salary:", highest_salary(employees))
```

```python
def highest_paid_employee(employees):
    highest = employees[0]

    for employee in employees:
        if employee["salary"] > highest["salary"]:
            highest = employee

    return highest

result = highest_paid_employee(employees)

print(result)
```

```python
def add_number(numbers, number):
    numbers.append(number)

numbers = [10, 20, 30]

add_number(numbers, 40)

print(numbers)
```

```python
def double_numbers(numbers):
    result = []

    for number in numbers:
        result.append(number * 2)

    return result

numbers = [1, 2, 3, 4]

new_numbers = double_numbers(numbers)

print(new_numbers)
```

# 📊 Data Analytics Example

Let's create a reusable function for sales analysis.

The function will accept a list of sales values and return
important business metrics.

```python
def sales_analysis(sales):
    total = sum(sales)
    average = total / len(sales)
    maximum = max(sales)
    minimum = min(sales)

    return {
        "Total Sales": total,
        "Average Sales": average,
        "Highest Sale": maximum,
        "Lowest Sale": minimum
    }

sales = [12000, 15000, 18000, 11000, 20000]

result = sales_analysis(sales)

print(result)
```

```python
def customer_summary(customer):
    return {
        "Name": customer["name"],
        "Total Purchases": sum(customer["purchases"]),
        "Average Purchase": sum(customer["purchases"]) / len(customer["purchases"])
    }

customer = {
    "name": "Priya",
    "purchases": [500, 800, 1200, 700]
}

result = customer_summary(customer)

print(result)
```
