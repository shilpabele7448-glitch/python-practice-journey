## What is Scope?

Scope means the area of a program where a variable can be accessed.

In simple words:

Scope tells us **where a variable is available**.

The two important scopes we will learn are:

- Local Scope
- Global Scope

## Local Variable

A variable created inside a function is called a local variable.

It can normally be accessed only inside that function.

```python
def student():
    name = "Rahul"
    print(name)

student()
```

```python
#NameError
def student():
    name = "Rahul"
    print(name)

student()

print(name)
```

```python
def calculate():
    total = 5000
    print("Total:", total)

calculate()
```

## Global Variable

A variable created outside a function is called a global variable.

A global variable can generally be accessed both:

- Outside the function
- Inside the function

```python
name = "Shilpa"

def greet():
    print("Hello", name)

greet()

print(name)
```

```python
company = "ABC Company"

def employee():
    print("Company:", company)

employee()
```

## Local vs Global Variables

| Local Variable | Global Variable |
|---|---|
| Created inside a function | Created outside a function |
| Available inside the function | Available throughout the program |
| Usually cannot be accessed outside the function | Can generally be accessed inside functions |
| Exists within its local scope | Exists in the global scope |

```python
name = "Rahul"

def student():
    name = "Amit"
    print("Inside:", name)

student()

print("Outside:", name)
```

Python checks for a variable in the local scope first.

Example:

name = "Rahul"

def student():
    name = "Amit"
    print(name)

Inside the function:

Python finds:
name = "Amit"

So it prints:

Amit

Outside the function:

Python finds:
name = "Rahul"

So it prints:

Rahul

```python
def greet(name):
    print("Hello", name)

greet("Rahul")
```

## The `global` Keyword

Normally, if we create a variable inside a function,
Python treats it as a local variable.

The `global` keyword tells Python that we want to work
with the global variable.

```python
count = 10

def update():
    count = 20
    print(count)

update()

print(count)
```

```python
count = 10

def update():
    global count
    count = 20

update()

print(count)
```

```python
score = 50

def increase_score():
    global score
    score += 10

increase_score()

print(score)
```

## ⚠️ Important

Although the `global` keyword exists, we should avoid using
global variables unnecessarily.

For larger programs, it is usually better to:

- Pass values as arguments
- Return values from functions

This makes the code easier to understand and maintain.

```python
def first():
    number = 10
    print("First:", number)

def second():
    number = 20
    print("Second:", number)

first()
second()
```

## 📊 Data Analytics Example

Suppose we have sales data.

We can create local variables inside a function
to perform calculations.

```python
sales = [10000, 15000, 12000, 18000]

def sales_analysis(sales):
    total = sum(sales)
    average = total / len(sales)

    print("Total Sales:", total)
    print("Average Sales:", average)

sales_analysis(sales)
```

```python
def sales_analysis(sales):
    total = sum(sales)
    average = total / len(sales)

    return total, average

sales = [10000, 15000, 12000, 18000]

total, average = sales_analysis(sales)

print("Total Sales:", total)
print("Average Sales:", average)
```

```python
x = 100

def test():
    x = 50
    print("Inside:", x)

test()

print("Outside:", x)
```

# 🧠 Quick Revision

### Local Variable
Created inside a function.

Example:

def test():
    x = 10

`x` is local.

---

### Global Variable
Created outside a function.

Example:

x = 10

def test():
    print(x)

---

### `global`
Used when we want to modify a global variable
inside a function.

Example:

x = 10

def test():
    global x
    x = 20

---

### Important Rule

Inside function:
    Local variable has priority.

Outside function:
    Global variable is available.
