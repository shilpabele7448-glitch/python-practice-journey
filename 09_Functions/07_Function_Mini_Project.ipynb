# 🐍 Employee Data Analysis Mini Project

This project analyzes employee data using Python functions.

The project will calculate:

- Total number of employees
- Total salary
- Average salary
- Highest salary
- Lowest salary
- Highest-paid employee
- Employees earning above average salary

```python
employees = [
    {"name": "Amit", "department": "IT", "salary": 50000},
    {"name": "Rahul", "department": "HR", "salary": 60000},
    {"name": "Priya", "department": "Finance", "salary": 55000},
    {"name": "Sneha", "department": "IT", "salary": 70000},
    {"name": "Rohan", "department": "Sales", "salary": 45000}
]

print(employees)
```

```python
def count_employees(employees):
    return len(employees)


total_employees = count_employees(employees)

print("Total Employees:", total_employees)
```

```python
def calculate_total_salary(employees):
    total = 0

    for employee in employees:
        total += employee["salary"]

    return total


total_salary = calculate_total_salary(employees)

print("Total Salary:", total_salary)
```

```python
def calculate_average_salary(employees):
    total = calculate_total_salary(employees)
    average = total / len(employees)

    return average


average_salary = calculate_average_salary(employees)

print("Average Salary:", average_salary)
```

```python
def find_highest_salary(employees):
    highest = employees[0]["salary"]

    for employee in employees:
        if employee["salary"] > highest:
            highest = employee["salary"]

    return highest


highest_salary = find_highest_salary(employees)

print("Highest Salary:", highest_salary)
```

```python
def find_lowest_salary(employees):
    lowest = employees[0]["salary"]

    for employee in employees:
        if employee["salary"] < lowest:
            lowest = employee["salary"]

    return lowest


lowest_salary = find_lowest_salary(employees)

print("Lowest Salary:", lowest_salary)
```

```python
def highest_paid_employee(employees):
    highest_employee = employees[0]

    for employee in employees:
        if employee["salary"] > highest_employee["salary"]:
            highest_employee = employee

    return highest_employee


result = highest_paid_employee(employees)

print("Highest-Paid Employee:", result["name"])
print("Department:", result["department"])
print("Salary:", result["salary"])
```

```python
def above_average_employees(employees):
    average = calculate_average_salary(employees)

    result = []

    for employee in employees:
        if employee["salary"] > average:
            result.append(employee)

    return result


above_average = above_average_employees(employees)

print(above_average)
```

```python
def department_count(employees):
    departments = {}

    for employee in employees:
        department = employee["department"]

        if department in departments:
            departments[department] += 1
        else:
            departments[department] = 1

    return departments


result = department_count(employees)

print(result)
```

```python
def employee_summary(employees):
    total_employees = count_employees(employees)
    total_salary = calculate_total_salary(employees)
    average_salary = calculate_average_salary(employees)
    highest_salary = find_highest_salary(employees)
    lowest_salary = find_lowest_salary(employees)
    highest_employee = highest_paid_employee(employees)

    return {
        "Total Employees": total_employees,
        "Total Salary": total_salary,
        "Average Salary": average_salary,
        "Highest Salary": highest_salary,
        "Lowest Salary": lowest_salary,
        "Highest-Paid Employee": highest_employee["name"]
    }


summary = employee_summary(employees)

print(summary)
```

```python
summary = employee_summary(employees)

for key, value in summary.items():
    print(key, ":", value)
```
