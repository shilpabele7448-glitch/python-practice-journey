## Return Statement

`return` sends a value back from the function.

Example:

```python

    return a + b

````python
def add(a, b):
    return a + b
````

````python
def add_print(a, b):
    print(a + b)
````

````python
def multiply(a, b):
    return a * b

result = multiply(5, 4)

print(result)
print(result + 10)
print(result * 2)
````

## Even or Odd Function

Create a function that checks whether a number is even or odd.

````python
def check_even_odd(number):
    
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"
````

````python
def square(number):
    return number * number
````

````python
def average(a, b, c):
    return (a + b + c) / 3
````

````python
def calculate_total(numbers):
    
    total = 0
    
    for number in numbers:
        total += number
    
    return total
````

## Data Analytics Example

Suppose we have sales data.

We can create a function that calculates total sales.

````python
def calculate_total_sales(sales):
    
    total = 0
    
    for sale in sales:
        total += sale
    
    return total
````

````python
def calculate_average_sales(sales):
    
    total = 0
    
    for sale in sales:
        total += sale
    
    return total / len(sales)
````
