## Positional Arguments

When values are passed to a function based on their position,
they are called positional arguments.

```python
def introduce(name, age):
    print("Name:", name)
    print("Age:", age)

introduce("Shilpa", 25)
```

```python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student("Rahul", 22)
```

## Keyword Arguments

Instead of relying on position, we can specify the parameter name.

This is called a keyword argument.

```python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student(age=22, name="Rahul")
```

```python
def calculate_total(price, quantity):
    return price * quantity

print(calculate_total(100, 5))
```

## Default Arguments

A function parameter can have a default value.

If the user doesn't provide a value,
Python uses the default value.

```python
def greet(name, message="Good Morning"):
    print(message, name)

greet("Shilpa")
```

```python
def student(name, age=18):
    print(name, age)
```

```python
def employee(name, department, salary):
    print("Name:", name)
    print("Department:", department)
    print("Salary:", salary)

employee("Amit", department="IT", salary=50000)
```

## *args

Sometimes we don't know how many positional arguments
the user will provide.

`*args` allows a function to accept multiple positional arguments.

Inside the function, `args` behaves like a tuple.

```python
def add_numbers(*args):
    print(args)

add_numbers(10, 20, 30, 40)
```

```python
def add_numbers(*args):
    total = 0

    for number in args:
        total += number

    return total

print(add_numbers(10, 20))
print(add_numbers(10, 20, 30))
print(add_numbers(10, 20, 30, 40, 50))
```

## **kwargs

`**kwargs` allows a function to accept multiple keyword arguments.

Inside the function, `kwargs` behaves like a dictionary.

```python
def student_info(**kwargs):
    print(kwargs)

student_info(name="Rahul", age=22, city="Pune")
```

```python
def student_info(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

student_info(
    name="Rahul",
    age=22,
    city="Pune"
)
```

## *args vs **kwargs

| Feature | *args | **kwargs |
|---|---|---|
| Accepts | Multiple positional arguments | Multiple keyword arguments |
| Stored as | Tuple | Dictionary |
| Example | `10, 20, 30` | `name="Rahul"` |
| Useful when | Number of values is unknown | Number of named values is unknown |

## Argument Unpacking

We can use `*` to unpack a list or tuple
when passing values to a function.

```python
def add(a, b, c):
    return a + b + c

numbers = [10, 20, 30]

print(add(*numbers))
```

```python
def student(name, age, city):
    print(name)
    print(age)
    print(city)

data = {
    "name": "Rahul",
    "age": 22,
    "city": "Pune"
}

student(**data)
```

```python
# common error
def add(a, b):
    return a + b

add(10, 20, 30)
```

## Data Analytics Example

Functions can make repeated calculations easier.

Suppose we have sales values for several days.

```python
def total_sales(*sales):
    return sum(sales)

sales = [12000, 15000, 18000, 11000, 20000]

print("Total Sales:", total_sales(*sales))
```

```python
def sales_summary(*sales):
    total = sum(sales)
    average = total / len(sales)
    maximum = max(sales)
    minimum = min(sales)

    return total, average, maximum, minimum

result = sales_summary(12000, 15000, 18000, 11000, 20000)

print("Total:", result[0])
print("Average:", result[1])
print("Maximum:", result[2])
print("Minimum:", result[3])
```
