## What is `return`?

The `return` statement sends a value from a function back
to the place where the function was called.

In simple words:

Function → performs calculation → returns result

```python
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
```

```python
def add_print(a, b):
    print(a + b)

add_print(10, 20)
```

## `print()` vs `return`

### print()

Displays a value on the screen.

### return

Sends a value back from the function.

Example:

def add(a, b):
    return a + b

result = add(10, 20)

Now we can use `result` again.

For example:

result * 2
result + 100
print(result)

```python
def square(number):
    return number * number

result = square(5)

print(result)
print(result + 10)
print(result * 2)
```

```python
def greet(name):
    return "Hello " + name

message = greet("Shilpa")

print(message)
```

```python
def is_even(number):
    return number % 2 == 0

result = is_even(10)

print(result)
```

```python
def get_numbers():
    return [10, 20, 30, 40, 50]

numbers = get_numbers()

print(numbers)
```

## Returning Multiple Values

Python allows a function to return multiple values.

Example:

return total, average

```python
def calculate(a, b):
    total = a + b
    difference = a - b

    return total, difference

result = calculate(20, 5)

print(result)
```

```python
def calculate(a, b):
    total = a + b
    difference = a - b

    return total, difference

total, difference = calculate(20, 5)

print("Total:", total)
print("Difference:", difference)
```

```python
def calculate(a, b):
    total = a + b
    average = (a + b) / 2
    product = a * b

    return total, average, product

total, average, product = calculate(10, 20)

print("Total:", total)
print("Average:", average)
print("Product:", product)
```

```python
def test():
    print("Statement 1")
    return
    print("Statement 2")

test()
```

```python
def check_number(number):
    if number > 0:
        return "Positive"
    else:
        return "Negative"

print(check_number(10))
print(check_number(-5))
```

```python
def check_number(number):
    if number > 0:
        return "Positive"
    elif number < 0:
        return "Negative"
    else:
        return "Zero"

print(check_number(10))
print(check_number(-5))
print(check_number(0))
```

```python
def add(a, b):
    print(a + b)

result = add(10, 20)

print("Result:", result)
```

```python
def square(number):
    return number * number

def double(number):
    return number * 2

result = double(square(5))

print(result)
```

```python
def calculate_total(price, quantity):
    return price * quantity

def add_tax(amount):
    return amount * 1.18

total = calculate_total(1000, 2)

final_amount = add_tax(total)

print("Total:", total)
print("Final Amount:", final_amount)
```

```python
def total_sales(sales):
    return sum(sales)

sales = [10000, 15000, 12000, 18000]

total = total_sales(sales)

print("Total Sales:", total)
```

```python
def sales_summary(sales):
    total = sum(sales)
    average = total / len(sales)
    highest = max(sales)
    lowest = min(sales)

    return total, average, highest, lowest

sales = [10000, 15000, 12000, 18000]

total, average, highest, lowest = sales_summary(sales)

print("Total Sales:", total)
print("Average Sales:", average)
print("Highest Sale:", highest)
print("Lowest Sale:", lowest)
```

```python
def student_result(marks):
    total = sum(marks)
    percentage = total / len(marks)

    return total, percentage

marks = [75, 82, 68, 91, 77]

total, percentage = student_result(marks)

print("Total:", total)
print("Percentage:", percentage)
```

```python
def sales_summary(sales):
    return {
        "Total": sum(sales),
        "Average": sum(sales) / len(sales),
        "Maximum": max(sales),
        "Minimum": min(sales)
    }

sales = [10000, 15000, 12000, 18000]

result = sales_summary(sales)

print(result)
```

# 🧠 Important Function Pattern

A very useful pattern to remember is:

1. Take input
2. Process the input
3. Return the result

Example:

def calculate_average(numbers):

    total = sum(numbers)
    average = total / len(numbers)

    return average

Input
  ↓
Process
  ↓
Return
  ↓
Use result
