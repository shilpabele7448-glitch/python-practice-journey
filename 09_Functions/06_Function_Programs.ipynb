## Program 1: Check Positive, Negative or Zero

Create a function that accepts a number and returns:

- "Positive"
- "Negative"
- "Zero"

```python
def check_number(number):
    if number > 0:
        return "Positive"
    elif number < 0:
        return "Negative"
    else:
        return "Zero"


print(check_number(10))
print(check_number(-5))
print(check_number(0))
```

```python
def check_even_odd(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"


print(check_even_odd(10))
print(check_even_odd(7))
```

```python
def largest_two(a, b):
    if a > b:
        return a
    else:
        return b


print(largest_two(25, 40))
```

```python
def largest_three(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c


print(largest_three(25, 70, 45))
```

## Program 5: Factorial

Factorial of a number means multiplying all positive
integers from that number down to 1.

Example:

5! = 5 × 4 × 3 × 2 × 1
   = 120

```python
def factorial(number):
    result = 1

    for i in range(1, number + 1):
        result *= i

    return result


print(factorial(5))
```

## Program 6: Check Prime Number

A prime number has exactly two factors:

1. 1
2. The number itself

Examples:

2, 3, 5, 7, 11

```python
def is_prime(number):
    if number < 2:
        return False

    for i in range(2, number):
        if number % i == 0:
            return False

    return True


print(is_prime(7))
print(is_prime(10))
```

## Program 7: Check Palindrome

A palindrome reads the same forward and backward.

Examples:

madam
level
radar
121

```python
def is_palindrome(text):
    return text == text[::-1]


print(is_palindrome("madam"))
print(is_palindrome("python"))
```

```python
def reverse_string(text):
    return text[::-1]


print(reverse_string("Python"))
```

## Program 9: Count Vowels

Create a function that counts the number of vowels
in a string.

Vowels:

a, e, i, o, u

```python
def count_vowels(text):
    count = 0

    for character in text.lower():
        if character in "aeiou":
            count += 1

    return count


print(count_vowels("Data Analytics"))
```

```python
def character_frequency(text):
    frequency = {}

    for character in text:
        if character in frequency:
            frequency[character] += 1
        else:
            frequency[character] = 1

    return frequency


print(character_frequency("banana"))
```

```python
def list_sum(numbers):
    total = 0

    for number in numbers:
        total += number

    return total


numbers = [10, 20, 30, 40]

print(list_sum(numbers))
```

```python
def list_average(numbers):
    total = 0

    for number in numbers:
        total += number

    return total / len(numbers)


numbers = [10, 20, 30, 40]

print(list_average(numbers))
```

```python
def find_even_numbers(numbers):
    even_numbers = []

    for number in numbers:
        if number % 2 == 0:
            even_numbers.append(number)

    return even_numbers


numbers = [1, 2, 3, 4, 5, 6, 7, 8]

print(find_even_numbers(numbers))
```

```python
def remove_duplicates(numbers):
    return list(set(numbers))


numbers = [10, 20, 10, 30, 20, 40]

print(remove_duplicates(numbers))
```

## Program 15: Student Result

Create a function that accepts marks and returns:

- Total
- Percentage
- Result

Passing percentage = 40%

```python
def student_result(marks):
    total = sum(marks)
    percentage = total / len(marks)

    if percentage >= 40:
        result = "Pass"
    else:
        result = "Fail"

    return total, percentage, result


marks = [75, 82, 68, 91, 77]

total, percentage, result = student_result(marks)

print("Total:", total)
print("Percentage:", percentage)
print("Result:", result)
```

## Program 16: Employee Salary Analysis

Given a list of employee dictionaries,
calculate:

- Total salary
- Average salary
- Highest salary
- Lowest salary

```python
def salary_analysis(employees):
    salaries = []

    for employee in employees:
        salaries.append(employee["salary"])

    total = sum(salaries)
    average = total / len(salaries)
    highest = max(salaries)
    lowest = min(salaries)

    return total, average, highest, lowest


employees = [
    {"name": "Amit", "salary": 50000},
    {"name": "Rahul", "salary": 60000},
    {"name": "Priya", "salary": 55000}
]

total, average, highest, lowest = salary_analysis(employees)

print("Total Salary:", total)
print("Average Salary:", average)
print("Highest Salary:", highest)
print("Lowest Salary:", lowest)
```

## Program 17: Sales Analysis

Create a function that accepts sales data and returns:

- Total Sales
- Average Sales
- Highest Sale
- Lowest Sale

```python
def sales_analysis(sales):
    total = sum(sales)
    average = total / len(sales)
    highest = max(sales)
    lowest = min(sales)

    return total, average, highest, lowest


sales = [12000, 15000, 18000, 11000, 20000]

total, average, highest, lowest = sales_analysis(sales)

print("Total Sales:", total)
print("Average Sales:", average)
print("Highest Sale:", highest)
print("Lowest Sale:", lowest)
```
